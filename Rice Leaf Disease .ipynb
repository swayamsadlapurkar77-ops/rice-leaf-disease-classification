# Executive Summary

## Project Overview

The Rice Leaf Disease Detection System is an end-to-end Deep Learning based Computer Vision project developed to automatically identify rice leaf diseases from digital images.

Agricultural crop diseases are one of the major challenges affecting crop productivity and food security. Early and accurate disease detection helps farmers take preventive actions and reduce crop losses.

This project applies advanced Deep Learning techniques and Transfer Learning approaches to build an automated multi-class rice disease classification system.

---

## Objective

The primary objective of this project is to develop an Artificial Intelligence based system capable of:

- Detecting rice leaf diseases from images
- Classifying leaves into multiple disease categories
- Reducing dependency on manual disease inspection
- Providing fast and accurate disease predictions
- Creating a deployment-ready machine learning pipeline

The project uses the PRCP-1001 RiceLeaf dataset containing images of three major rice leaf diseases:

| Disease Class | Description |

# Problem Statement

## Background

Agriculture plays a critical role in global food production, and rice is one of the most important staple crops worldwide.

However, rice crops are highly affected by various diseases that reduce crop quality, productivity, and economic value for farmers.

Traditional disease identification methods rely on manual inspection by agricultural experts, which can be time-consuming, expensive, and difficult to scale.

---

## Problem Definition

The objective of this project is to develop an Artificial Intelligence based automated system that can identify rice leaf diseases from images using Deep Learning techniques.

The system should classify rice leaf images into different disease categories accurately and provide fast predictions to support early disease detection.

---

## Challenges Addressed

The project focuses on solving the following challenges:

- Manual disease detection requires expert knowledge

```python
import sys
import os
import random
import time
import logging
import tensorflow as tf
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import cv2
from zipfile import ZipFile
import requests
from sklearn.metrics import classification_report, confusion_matrix, roc_curve, auc, precision_recall_curve
from sklearn.preprocessing import label_binarize
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.applications import MobileNetV2, EfficientNetB0, ResNet50
from tensorflow.keras.layers import Dense, GlobalAveragePooling2D, Dropout
from tensorflow.keras.models import Model
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint, ReduceLROnPlateau, CSVLogger, TensorBoard
```

**1. Configuration Class** 

```python
class Config:
    SEED = 42
    IMG_SIZE = (224, 224)
    BATCH_SIZE = 32
    INITIAL_EPOCHS = 10
    FINE_TUNE_EPOCHS = 10
    SELECTED_MODEL = "MobileNetV2" 
    
   
    DIR_DATA = "dataset"
    DIR_MODELS = "models"
    DIR_LOGS = "logs"
    DIR_TENSORBOARD = "logs/tensorboard"
    DIR_ARTIFACTS = "artifacts"
    DIR_REPORTS = "reports"
    
    DATA_URL = "https://d3ilbtxij3aepc.cloudfront.net/projects/CDS-Capstone-Projects/PRCP-1001-RiceLeaf.zip"
    ZIP_PATH = "rice_leaf.zip"


for directory in [Config.DIR_DATA, Config.DIR_MODELS, Config.DIR_LOGS, Config.DIR_TENSORBOARD, Config.DIR_ARTIFACTS, Config.DIR_REPORTS]:
    os.makedirs(directory, exist_ok=True)


logging.basicConfig(level=logging.INFO, format="%(asctime)s - %(levelname)s - %(message)s")
```

**2. Global Random Seed Enforcement**

```python
os.environ['PYTHONHASHSEED'] = str(Config.SEED)
random.seed(Config.SEED)
np.random.seed(Config.SEED)
tf.random.set_seed(Config.SEED)

print("="*50)
print(" SYSTEM, HARDWARE & SEED DETAILS")
print("="*50)
print(f"Python Version: {sys.version.split(' ')[0]}")
print(f"TensorFlow Version: {tf.__version__}")
print(f"Global Seed Set To: {Config.SEED}")
```

**3. GPU Memory Growth Setup**

```python
gpus = tf.config.list_physical_devices('GPU')
if gpus:
    print(f"GPU Available: YES ({len(gpus)} detected)")
    try:
        for gpu in gpus:
            tf.config.experimental.set_memory_growth(gpu,True)
            print(f" - {gpu.name} (Menmory Growth: ENABLED)")
    except RuntimeError as e:
        print(f"GPU Error:{e}")
else:
    print("GPU Available: NO (Running on CPU)")
print("="*50)
```

```python
import os

def get_dataset_path(base_dir):
    for root, dirs, files in os.walk(base_dir):
        
        class_dirs = [d for d in dirs if not d.startswith('.')]
        
        if len(class_dirs) >= 2:
            return root
    
    return base_dir 


DATASET_PATH = get_dataset_path("dataset")

print("FINAL DATASET PATH:", DATASET_PATH)
```

```python
import os
import shutil
import logging

logging.basicConfig(level=logging.INFO)

source = "dataset/Data"
destination = "dataset"

def reorganize_dataset(source, destination):

    # Check source folder
    if not os.path.exists(source):
        logging.info("Dataset structure already organized. Skipping reorganization.")
        return

    moved_count = 0

    for folder in os.listdir(source):

        folder_path = os.path.join(source, folder)

        if os.path.isdir(folder_path):

            inner_items = os.listdir(folder_path)

            # Handle nested folders
            if len(inner_items) == 1:

                class_folder = os.path.join(folder_path, inner_items[0])

                if os.path.isdir(class_folder):

                    new_path = os.path.join(destination, inner_items[0])

                    if not os.path.exists(new_path):
                        shutil.move(class_folder, new_path)
                        moved_count += 1
                        logging.info(f"Moved class folder: {inner_items[0]}")

    if moved_count > 0:
        logging.info(f"Dataset reorganized successfully. {moved_count} folders moved.")
    else:
        logging.info("No changes required.")

        
reorganize_dataset(source, destination)
```

```python
import os

DATASET_DIR = "dataset"

print("\nDATASET VALIDATION START\n" + "="*50)


if not os.path.exists(DATASET_DIR):
    raise FileNotFoundError(f"Dataset folder not found: {DATASET_DIR}")


def count_images(folder):
    count = 0
    for root, dirs, files in os.walk(folder):
        for f in files:
            if f.lower().endswith((".jpg", ".jpeg", ".png")):
                count += 1
    return count


classes = []
for d in os.listdir(DATASET_DIR):
    path = os.path.join(DATASET_DIR, d)

    if os.path.isdir(path):
     
        if d.lower() not in ["data", "dataset", "__pycache__"]:
            classes.append(d)

print(f"✔ Dataset Path: {DATASET_DIR}")
print(f"✔ Classes Found: {classes}")


class_counts = {}

for cls in classes:
    class_path = os.path.join(DATASET_DIR, cls)
    class_counts[cls] = count_images(class_path)

# Print distribution
print("\n📊 CLASS DISTRIBUTION:")
for cls, count in class_counts.items():
    print(f"{cls}: {count} images")


valid_counts = [c for c in class_counts.values() if c > 0]

if len(valid_counts) == 0:
    print("\n❌ Dataset is Empty or Invalid")
else:
    imbalance_ratio = max(valid_counts) / min(valid_counts)

    print("\n⚖️ IMBALANCE RATIO:", round(imbalance_ratio, 2))

    if imbalance_ratio > 1.5:
        print("⚠️ Dataset is Imbalanced")
    else:
        print("✅ Dataset is Balanced")

print("\n" + "="*50)
print("✅ DATASET VALIDATION COMPLETE")
```

**4. Data Augmentation**

```python
import matplotlib.pyplot as plt
import numpy as np
import tensorflow as tf

def visualize_augmented_samples(generator, num_images=6, seed=42):

    class_names = list(generator.class_indices.keys())

    # IMPORTANT: reset generator
    generator.reset()

    # fix randomness for visualization ONLY
    np.random.seed(seed)
    tf.random.set_seed(seed)

    plt.figure(figsize=(10,6))

    for i in range(num_images):
        img, label = next(generator)

        class_index = np.argmax(label[0])
        class_name = class_names[class_index]

        plt.subplot(2,3,i+1)
        plt.imshow(img[0])
        plt.title(class_name)
        plt.axis("off")

    plt.suptitle("DATA AUGMENTATION VISUALIZATION (REPRODUCIBLE VIEW)")
    plt.tight_layout()
    plt.show()

# CALL
visualize_augmented_samples(train_gen)
```

```python
import matplotlib.pyplot as plt
import numpy as np

def plot_class_distribution(generator, title="Class Distribution"):
    labels = generator.classes
    class_names = list(generator.class_indices.keys())

    
    counts = np.bincount(labels)

    plt.figure(figsize=(8,5))
    bars = plt.bar(class_names, counts, color='skyblue')

    plt.title(title)
    plt.xlabel("Classes")
    plt.ylabel("Number of Images")

    
    for bar in bars:
        height = bar.get_height()
        plt.text(bar.get_x() + bar.get_width()/2, height,
                 f'{int(height)}',
                 ha='center', va='bottom')

    plt.xticks(rotation=20)
    plt.tight_layout()
    plt.show()


plot_class_distribution(train_gen)
```

# 4.**BASELINE CNN MODEL DEVELOPMENT**

**IMPLEMENTS:-**

**1. Import Required Libraries**

import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt
import os

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import (
    Conv2D,
    MaxPooling2D,
    BatchNormalization,
    Flatten,
    Dense,
    Dropout
)

from tensorflow.keras.optimizers import Adam

from tensorflow.keras.callbacks import (
    EarlyStopping,
    ModelCheckpoint,
    ReduceLROnPlateau,
    CSVLogger
)

**2.CNN Configuration**

```python
class Config:

    
    DATASET_PATH = "dataset"

   
    SEED = 42

   
    IMG_SIZE = (224,224)
    CHANNELS = 3

   
    BATCH_SIZE = 16
    EPOCHS = 30

    
    NUM_CLASSES = 3

    
    LEARNING_RATE = 0.001

    
    MODEL_DIR = "models"
    CNN_MODEL_PATH = "models/baseline_cnn.keras"

    
   
```

**3.Dataset Pipeline Construction**

```python
train_ds = tf.keras.utils.image_dataset_from_directory(
    Config.DATASET_PATH,
    validation_split=0.2,
    subset="training",
    seed=Config.SEED,
    image_size=Config.IMG_SIZE,
    batch_size=Config.BATCH_SIZE
)


val_ds = tf.keras.utils.image_dataset_from_directory(
    Config.DATASET_PATH,
    validation_split=0.2,
    subset="validation",
    seed=Config.SEED,
    image_size=Config.IMG_SIZE,
    batch_size=Config.BATCH_SIZE
)
```

**4.Dataset Class Verification**

```python
class_names = train_ds.class_names

print("Classes:")
print(class_names)
```

**5.Image Preprocessing Pipeline**

```python
normalization_layer = tf.keras.layers.Rescaling(1./255)

train_ds = train_ds.map(
    lambda x,y:(normalization_layer(x),y)
)

val_ds = val_ds.map(
    lambda x,y:(normalization_layer(x),y)
)
```

**6.Data Pipeline Optimization**

```python
AUTOTUNE = tf.data.AUTOTUNE

train_ds = (
    train_ds
    .cache()
    .shuffle(1000)
    .prefetch(AUTOTUNE)
)

val_ds = (
    val_ds
    .cache()
    .prefetch(AUTOTUNE)
)
```

**7.Baseline CNN Architecture Design**

```python
from tensorflow.keras.models import Sequential

from tensorflow.keras.layers import (
    Conv2D,
    MaxPooling2D,
    Flatten,
    Dense,
    Dropout,
    BatchNormalization
)
```

```python
cnn_model = Sequential([


    Conv2D(
        32,
        (3,3),
        activation="relu",
        input_shape=(224,224,3)
    ),

    BatchNormalization(),

    MaxPooling2D(),



    Conv2D(
        64,
        (3,3),
        activation="relu"
    ),

    BatchNormalization(),

    MaxPooling2D(),



    Conv2D(
        128,
        (3,3),
        activation="relu"
    ),

    BatchNormalization(),

    MaxPooling2D(),



    Flatten(),


    Dense(
        128,
        activation="relu"
    ),


    Dropout(0.5),


    Dense(
        Config.NUM_CLASSES,
        activation="softmax"
    )

])
```

**8.Model Summary**

```python
cnn_model.summary()
```

**9.Model Compilation**

```python
cnn_model.compile(

    optimizer=Adam(
        learning_rate=Config.LEARNING_RATE
    ),

    loss="sparse_categorical_crossentropy",

    metrics=[
        "accuracy"
    ]

)
```

**10.Training Control & Experiment Monitoring**

```python
os.makedirs(
    Config.MODEL_DIR,
    exist_ok=True
)


callbacks = [

    EarlyStopping(
        monitor="val_loss",
        patience=5,
        restore_best_weights=True
    ),


    ModelCheckpoint(
        Config.CNN_MODEL_PATH,
        monitor="val_accuracy",
        save_best_only=True
    ),


    ReduceLROnPlateau(
        monitor="val_loss",
        factor=0.2,
        patience=3
    ),


    CSVLogger(
        "cnn_training_log.csv"
    )

]
```

**11. CNN Model Training Execution**


```python
history_cnn = cnn_model.fit(

    train_ds,

    validation_data=val_ds,

    epochs=Config.EPOCHS,

    callbacks=callbacks

)
```

**12.Training Performance Analysis**

**Accuracy Curve**

* Training accuracy
* Validation accuracy


```python
plt.figure(figsize=(8,5))

plt.plot(
    history_cnn.history["accuracy"],
    label="Train Accuracy"
)

plt.plot(
    history_cnn.history["val_accuracy"],
    label="Validation Accuracy"
)


plt.title("Baseline CNN Accuracy")

plt.xlabel("Epoch")

plt.ylabel("Accuracy")

plt.legend()

plt.show()
```

**Loss Curve**

* Training loss
* Validation loss

```python
plt.figure(figsize=(8,5))


plt.plot(
    history_cnn.history["loss"],
    label="Train Loss"
)


plt.plot(
    history_cnn.history["val_loss"],
    label="Validation Loss"
)


plt.title("Baseline CNN Loss")

plt.xlabel("Epoch")

plt.ylabel("Loss")

plt.legend()

plt.show()
```

**13.Baseline CNN Evaluation**





```python
loss, accuracy = cnn_model.evaluate(val_ds)

print(
    f"Validation Accuracy: {accuracy*100:.2f}%"
)
```

## 5.MobileNetV2 Transfer Learning ##

**1.Experiment Configuration**

* Implements:-This section centralizes all hyperparameters and settings used for the MobileNetV2 experiment. Keeping these in one place makes the experiment reproducible and easier to modify.

```python
import tensorflow as tf
import numpy as np
import random
import os


# ============================================================
# Reproducibility Configuration
# ============================================================

SEED = 42

tf.random.set_seed(SEED)
np.random.seed(SEED)
random.seed(SEED)


# ============================================================
# MobileNetV2 Configuration
# ============================================================

MOBILENET_CONFIG = {

    "IMG_SIZE": (224,224),

    "BATCH_SIZE": 32,

    "INITIAL_EPOCHS": 10,

    "FINE_TUNE_EPOCHS": 10,


    # Phase 1 Training
    "INITIAL_LEARNING_RATE": 1e-4,


    # Phase 2 Fine Tuning
    "FINE_TUNE_LEARNING_RATE": 1e-5,


    "DROPOUT_RATE": 0.30,

    "DENSE_UNITS": 256,


    "UNFREEZE_PERCENTAGE": 0.20,


    "NUM_CLASSES": NUM_CLASSES,


    "MODEL_NAME": "MobileNetV2"

}
```

**2.Load MobileNetV2 Pretrained Backbone**

* Implements :


```python


import logging
from tensorflow.keras.applications import MobileNetV2



# Logger Configuration


logger = logging.getLogger("RiceLeaf_MobileNetV2")

if not logger.handlers:
    logging.basicConfig(
        level=logging.INFO,
        format="%(asctime)s | %(levelname)s | %(message)s"
    )



# Load MobileNetV2 Backbone


try:

    logger.info("=" * 70)
    logger.info("Initializing MobileNetV2 pretrained backbone")
    logger.info("=" * 70)


    mobilenet_base = MobileNetV2(
        weights="imagenet",
        include_top=False,
        input_shape=MOBILENET_CONFIG["IMG_SIZE"] + (3,)
    )


    # Model Metadata
    model_info = {
        "Model": "MobileNetV2",
        "Weights": "ImageNet",
        "Include Top": False,
        "Input Shape": mobilenet_base.input_shape,
        "Output Shape": mobilenet_base.output_shape,
        "Total Layers": len(mobilenet_base.layers),
        "Trainable Layers": sum(
            layer.trainable 
            for layer in mobilenet_base.layers
        )
    }


    logger.info("MobileNetV2 backbone loaded successfully")


    for key, value in model_info.items():
        logger.info(f"{key}: {value}")


    logger.info("=" * 70)


except Exception as e:

    logger.exception(
        f"Failed to load MobileNetV2 backbone: {str(e)}"
    )

    raise e
```

```python
# MobileNetV2 architecture summary
mobilenet_base.summary()
```

**3.Build Transfer Learning Architecture**

* Implements:
* Use MobileNetV2 as feature extractor
  * Add custom classification head

* Global Average Pooling
* Batch Normalization
* Dropout (regularization)
* Dense classifier
* Softmax output layer
* Use Functional API (industry standard)
* Add logging
* Add exception handling
* Validate architecture

```python



import tensorflow as tf
from tensorflow.keras import layers, Model
import logging


logger = logging.getLogger("RiceLeaf_MobileNetV2")


try:

    logger.info("=" * 70)
    logger.info("Building MobileNetV2 Transfer Learning Architecture")
    logger.info("=" * 70)


   
    # Input Layer
   

    inputs = tf.keras.Input(
        shape=MOBILENET_CONFIG["IMG_SIZE"] + (3,),
        name="input_image"
    )


    
    # Feature Extraction Backbone
    

    x = mobilenet_base(
        inputs,
        training=False
    )


    logger.info(
        f"Feature extractor output shape: {x.shape}"
    )


    # Classification Head
    

    x = layers.GlobalAveragePooling2D(
        name="global_average_pooling"
    )(x)


    x = layers.BatchNormalization(
        name="batch_normalization"
    )(x)


    x = layers.Dropout(
        rate=0.3,
        name="dropout_regularization"
    )(x)


    x = layers.Dense(
        128,
        activation="relu",
        name="feature_dense_layer"
    )(x)


    x = layers.Dropout(
        rate=0.2,
        name="classifier_dropout"
    )(x)


    outputs = layers.Dense(
        MOBILENET_CONFIG["NUM_CLASSES"],
        activation="softmax",
        name="classification_output"
    )(x)



   
    # Create Final Model
   

    mobilenet_model = Model(
        inputs,
        outputs,
        name="RiceLeaf_MobileNetV2_TransferLearning"
    )


    
    # Model Summary Information
  

    total_params = mobilenet_model.count_params()

    trainable_params = sum(
        tf.keras.backend.count_params(w)
        for w in mobilenet_model.trainable_weights
    )

    non_trainable_params = sum(
        tf.keras.backend.count_params(w)
        for w in mobilenet_model.non_trainable_weights
    )


    logger.info("Model architecture created successfully")

    logger.info(f"Model Name              : {mobilenet_model.name}")
    logger.info(f"Total Parameters        : {total_params:,}")
    logger.info(f"Trainable Parameters    : {trainable_params:,}")
    logger.info(f"Non Trainable Parameters: {non_trainable_params:,}")


    logger.info("=" * 70)


except Exception as e:

    logger.exception(
        f"Architecture building failed: {str(e)}"
    )

    raise e
                    
```

**4.Freeze Feature Extractor**
 * first phase of transfer learning, we do not train the ImageNet feature extractor.
 
Reason:

* MobileNetV2 already learned powerful visual features from ImageNet.
* Your dataset is small (Rice Leaf Disease dataset).
* Training all layers initially can cause:
* Overfitting
* Loss of pretrained knowledge
* Unstable training

```python


import logging
import tensorflow as tf


logger = logging.getLogger("RiceLeaf_MobileNetV2")


try:

    logger.info("=" * 70)
    logger.info("Freezing MobileNetV2 Feature Extractor")
    logger.info("=" * 70)


    
    # Freeze MobileNetV2 Backbone
   

    mobilenet_base.trainable = False


    logger.info(
        "MobileNetV2 backbone frozen successfully"
    )



    # Verify Layer Status
    

    total_layers = len(mobilenet_base.layers)

    trainable_layers = sum(
        1 for layer in mobilenet_base.layers
        if layer.trainable
    )

    frozen_layers = total_layers - trainable_layers


    logger.info(
        f"Total Backbone Layers : {total_layers}"
    )

    logger.info(
        f"Trainable Backbone Layers : {trainable_layers}"
    )

    logger.info(
        f"Frozen Backbone Layers : {frozen_layers}"
    )


    
    # Recalculate Model Parameters


    total_params = mobilenet_model.count_params()


    trainable_params = sum(
        tf.keras.backend.count_params(weight)
        for weight in mobilenet_model.trainable_weights
    )


    non_trainable_params = sum(
        tf.keras.backend.count_params(weight)
        for weight in mobilenet_model.non_trainable_weights
    )


    logger.info("------------------------------------------")
    logger.info(
        f"Total Parameters         : {total_params:,}"
    )

    logger.info(
        f"Trainable Parameters     : {trainable_params:,}"
    )

    logger.info(
        f"Non Trainable Parameters : {non_trainable_params:,}"
    )


    logger.info("=" * 70)


except Exception as e:

    logger.exception(
        f"Feature extractor freezing failed: {str(e)}"
    )

    raise e
```

**5.Compile Initial Model**

* Implements :

* Optimizer → Adam
* Learning Rate → 1e-4
 * Loss → Categorical Crossentropy
* Metrics:
* Accuracy
* Precision
* Recall
* Logging
* Exception Handling

```python


import logging
import tensorflow as tf


logger = logging.getLogger("RiceLeaf_MobileNetV2")


try:

    logger.info("=" * 70)
    logger.info("Compiling Initial MobileNetV2 Transfer Learning Model")
    logger.info("=" * 70)


    
    # Optimizer Configuration
    

    optimizer = tf.keras.optimizers.Adam(
    learning_rate=MOBILENET_CONFIG["INITIAL_LEARNING_RATE"]

    )


   
    # Compile Model
   

    mobilenet_model.compile(

        optimizer=optimizer,

        loss=tf.keras.losses.CategoricalCrossentropy(),

        metrics=[
            tf.keras.metrics.CategoricalAccuracy(
                name="accuracy"
            ),

            tf.keras.metrics.Precision(
                name="precision"
            ),

            tf.keras.metrics.Recall(
                name="recall"
            )
        ]
    )


    logger.info(
        "Model compiled successfully"
    )


    
    # Compilation Metadata
    

    compile_info = {

        "Optimizer": "Adam",

        "Learning Rate":
    MOBILENET_CONFIG["INITIAL_LEARNING_RATE"],
        
        "Loss":
            "Categorical Crossentropy",

        "Metrics":
            [
                "Accuracy",
                "Precision",
                "Recall"
            ]
    }


    for key, value in compile_info.items():

        logger.info(
            f"{key}: {value}"
        )


    logger.info("=" * 70)


except Exception as e:

    logger.exception(
        f"Model compilation failed: {str(e)}"
    )

    raise e
```

**6.Configure Training Callbacks**

* Implements :

* Saving the best model automatically

* Preventing overfitting

* Adjusting learning rate dynamically

* Logging training history

* Monitoring experiments

```python


import os
import logging
from tensorflow.keras.callbacks import (
    ModelCheckpoint,
    EarlyStopping,
    ReduceLROnPlateau,
    CSVLogger,
    TensorBoard
)


logger = logging.getLogger("RiceLeaf_MobileNetV2")


try:

    logger.info("=" * 70)
    logger.info("Configuring Training Callbacks")
    logger.info("=" * 70)


    
    # Create Directory Structure
    

    CHECKPOINT_DIR = "artifacts/checkpoints"
    LOG_DIR = "artifacts/logs"
    HISTORY_DIR = "artifacts/history"


    os.makedirs(
        CHECKPOINT_DIR,
        exist_ok=True
    )

    os.makedirs(
        LOG_DIR,
        exist_ok=True
    )

    os.makedirs(
        HISTORY_DIR,
        exist_ok=True
    )


   
    # Model Checkpoint
    

    checkpoint_path = os.path.join(
        CHECKPOINT_DIR,
        "best_mobilenetv2_model.keras"
    )


    checkpoint_callback = ModelCheckpoint(

        filepath=checkpoint_path,

        monitor="val_accuracy",

        save_best_only=True,

        save_weights_only=False,

        mode="max",

        verbose=1
    )


    
    # Early Stopping
    

    early_stopping_callback = EarlyStopping(

        monitor="val_loss",

        patience=5,

        restore_best_weights=True,

        verbose=1
    )


    
    # Learning Rate Scheduler
    

    reduce_lr_callback = ReduceLROnPlateau(

        monitor="val_loss",

        factor=0.2,

        patience=3,

        min_lr=1e-7,

        verbose=1
    )


    
    # CSV Training Logger
 

    csv_logger_callback = CSVLogger(

        filename=os.path.join(
            HISTORY_DIR,
            "training_history.csv"
        ),

        append=True
    )


    
    # TensorBoard Logger
    

    tensorboard_callback = TensorBoard(

        log_dir=LOG_DIR,

        histogram_freq=1
    )


 
    # Combine Callbacks
   

    CALLBACKS = [

        checkpoint_callback,

        early_stopping_callback,

        reduce_lr_callback,

        csv_logger_callback,

        tensorboard_callback

    ]


    logger.info(
        "Training callbacks configured successfully"
    )


    logger.info("------------------------------------------")

    logger.info(
        f"Checkpoint Path: {checkpoint_path}"
    )

    logger.info(
        f"TensorBoard Log Directory: {LOG_DIR}"
    )

    logger.info(
        f"Number of Callbacks: {len(CALLBACKS)}"
    )


    logger.info("------------------------------------------")

    logger.info("=" * 70)



except Exception as e:

    logger.exception(
        f"Callback configuration failed: {str(e)}"
    )

    raise e
```

```python


import logging
import numpy as np
import matplotlib.pyplot as plt

from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    accuracy_score
)

import seaborn as sns


logger = logging.getLogger("RiceLeaf_MobileNetV2")


try:

    logger.info("=" * 70)
    logger.info("Evaluating Initial MobileNetV2 Model")
    logger.info("=" * 70)


    
    # Model Evaluation
  

    evaluation_results = mobilenet_model.evaluate(
        validation_generator,
        verbose=1
    )


    for metric, value in zip(
        mobilenet_model.metrics_names,
        evaluation_results
    ):

        logger.info(
            f"{metric}: {value:.4f}"
        )


    
    # Generate Predictions
  

    validation_generator.reset()


    predictions = mobilenet_model.predict(
        validation_generator,
        verbose=1
    )


    predicted_classes = np.argmax(
        predictions,
        axis=1
    )


    true_classes = validation_generator.classes


    
    # Accuracy Calculation
    

    accuracy = accuracy_score(
        true_classes,
        predicted_classes
    )


    logger.info(
        f"Calculated Validation Accuracy: {accuracy:.4f}"
    )


    
    # Classification Report
  

    class_names = list(
        validation_generator.class_indices.keys()
    )


    report = classification_report(
        true_classes,
        predicted_classes,
        target_names=class_names
    )


    logger.info(
        "\nClassification Report:\n"
        + report
    )


    print("\nClassification Report\n")
    print(report)


    
    # Confusion Matrix
    
    cm = confusion_matrix(
        true_classes,
        predicted_classes
    )


    plt.figure(
        figsize=(7,5)
    )


    sns.heatmap(
        cm,
        annot=True,
        fmt="d",
        xticklabels=class_names,
        yticklabels=class_names
    )


    plt.xlabel(
        "Predicted"
    )

    plt.ylabel(
        "Actual"
    )

    plt.title(
        "MobileNetV2 Initial Model Confusion Matrix"
    )

    plt.show()


    logger.info("=" * 70)
    logger.info("Initial Model Evaluation Completed")
    logger.info("=" * 70)



except Exception as e:

    logger.exception(
        f"Evaluation failed: {str(e)}"
    )

    raise e
```

**9.Fine-Tune MobileNetV2**

**Implements:**

```python


import logging
import tensorflow as tf


logger = logging.getLogger("RiceLeaf_MobileNetV2")


try:

    logger.info("=" * 70)
    logger.info("Starting MobileNetV2 Fine-Tuning Configuration")
    logger.info("=" * 70)


    # Unfreeze MobileNetV2 Backbone
    

    mobilenet_base.trainable = True


    total_layers = len(mobilenet_base.layers)


    # Calculate layers to freeze
    freeze_until = int(
        total_layers *
        (1 - MOBILENET_CONFIG["UNFREEZE_PERCENTAGE"])
    )


    logger.info(
        f"Total Backbone Layers: {total_layers}"
    )

    logger.info(
        f"Freezing first {freeze_until} layers"
    )


    
    # Freeze First 80%
    # Train Last 20%
   

    for layer in mobilenet_base.layers[:freeze_until]:

        layer.trainable = False


    for layer in mobilenet_base.layers[freeze_until:]:

        layer.trainable = True



   
    # Batch Normalization Handling
   

    # Keep BatchNorm frozen during fine tuning
    # Improves stability on small datasets

    for layer in mobilenet_base.layers:

        if isinstance(
            layer,
            tf.keras.layers.BatchNormalization
        ):

            layer.trainable = False



    
    # Layer Statistics
    

    trainable_layers = sum(
        1 for layer in mobilenet_base.layers
        if layer.trainable
    )


    frozen_layers = total_layers - trainable_layers


    logger.info("------------------------------------------")

    logger.info(
        f"Trainable Backbone Layers: {trainable_layers}"
    )

    logger.info(
        f"Frozen Backbone Layers: {frozen_layers}"
    )


    
    # Parameter Statistics
    

    total_params = mobilenet_model.count_params()


    trainable_params = sum(
        tf.keras.backend.count_params(w)
        for w in mobilenet_model.trainable_weights
    )


    non_trainable_params = sum(
        tf.keras.backend.count_params(w)
        for w in mobilenet_model.non_trainable_weights
    )


    logger.info("------------------------------------------")

    logger.info(
        f"Total Parameters: {total_params:,}"
    )

    logger.info(
        f"Trainable Parameters: {trainable_params:,}"
    )

    logger.info(
        f"Non Trainable Parameters: {non_trainable_params:,}"
    )


    logger.info("------------------------------------------")

    logger.info(
        "MobileNetV2 fine-tuning configuration completed"
    )

    logger.info("=" * 70)



except Exception as e:

    logger.exception(
        f"Fine-tuning configuration failed: {str(e)}"
    )

    raise e
```

```python


import logging
import tensorflow as tf


logger = logging.getLogger("RiceLeaf_MobileNetV2")


try:

    logger.info("=" * 70)
    logger.info("Recompiling Fine-Tuned MobileNetV2 Model")
    logger.info("=" * 70)


    
    # Fine-Tuning Optimizer


    fine_tune_optimizer = tf.keras.optimizers.Adam(

        learning_rate=
        MOBILENET_CONFIG["FINE_TUNE_LEARNING_RATE"]

    )


    
    # Recompile Model
  

    mobilenet_model.compile(

        optimizer=fine_tune_optimizer,

        loss=tf.keras.losses.CategoricalCrossentropy(),

        metrics=[

            tf.keras.metrics.CategoricalAccuracy(
                name="accuracy"
            ),

            tf.keras.metrics.Precision(
                name="precision"
            ),

            tf.keras.metrics.Recall(
                name="recall"
            )

        ]

    )


   
    # Parameter Statistics
   

    total_params = mobilenet_model.count_params()


    trainable_params = sum(

        tf.keras.backend.count_params(weight)

        for weight in mobilenet_model.trainable_weights

    )


    non_trainable_params = sum(

        tf.keras.backend.count_params(weight)

        for weight in mobilenet_model.non_trainable_weights

    )


    logger.info(
        "Fine-tuned model compiled successfully"
    )


    logger.info("------------------------------------------")


    logger.info(
        f"Optimizer              : Adam"
    )

    logger.info(
        f"Learning Rate          : "
        f"{MOBILENET_CONFIG['FINE_TUNE_LEARNING_RATE']}"
    )

    logger.info(
        f"Loss                   : Categorical Crossentropy"
    )

    logger.info(
        f"Total Parameters       : {total_params:,}"
    )

    logger.info(
        f"Trainable Parameters   : {trainable_params:,}"
    )

    logger.info(
        f"Non Trainable Params   : {non_trainable_params:,}"
    )


    logger.info("------------------------------------------")

    logger.info("=" * 70)



except Exception as e:

    logger.exception(
        f"Fine-tuned model compilation failed: {str(e)}"
    )

    raise e
```

**3.Build Transfer Learning Architecture**

**Implements**

```python


import logging
import tensorflow as tf

from tensorflow.keras import Model
from tensorflow.keras.layers import (
    Input,
    GlobalAveragePooling2D,
    BatchNormalization,
    Dense,
    Dropout
)

logger = logging.getLogger("RiceLeaf_EfficientNetB0")

try:

    logger.info("=" * 70)
    logger.info("Building EfficientNetB0 Transfer Learning Architecture")
    logger.info("=" * 70)

  
    # Input Layer
   

    inputs = Input(
        shape=EFFICIENTNET_CONFIG["IMG_SIZE"] + (3,),
        name="Input_Image"
    )

   
    # Feature Extraction
    

    x = efficientnet_base(
        inputs,
        training=False
    )

    logger.info(
        f"Feature extractor output shape: {x.shape}"
    )

    
    # Global Average Pooling
    

    x = GlobalAveragePooling2D(
        name="GlobalAveragePooling"
    )(x)

    
    # Batch Normalization
    

    x = BatchNormalization(
        name="BatchNormalization"
    )(x)

    
    # Fully Connected Layer
    

    x = Dense(

        EFFICIENTNET_CONFIG["DENSE_UNITS"],

        activation="relu",

        name="Dense_256"

    )(x)

    
    # Dropout
    

    x = Dropout(

        EFFICIENTNET_CONFIG["DROPOUT_RATE"],

        name="Dropout"

    )(x)

   
    # Output Layer
 

    outputs = Dense(

        EFFICIENTNET_CONFIG["NUM_CLASSES"],

        activation="softmax",

        name="Prediction"

    )(x)

   
    # Build Model
    

    efficientnet_model = Model(

        inputs=inputs,

        outputs=outputs,

        name="RiceLeaf_EfficientNetB0_TransferLearning"

    )

   
    # Model Statistics
    

    total_params = efficientnet_model.count_params()

    trainable_params = sum(

        tf.keras.backend.count_params(weight)

        for weight in efficientnet_model.trainable_weights

    )

    non_trainable_params = sum(

        tf.keras.backend.count_params(weight)

        for weight in efficientnet_model.non_trainable_weights

    )

    logger.info(
        "Model architecture created successfully"
    )

    logger.info("------------------------------------------")

    logger.info(
        f"Model Name              : {efficientnet_model.name}"
    )

    logger.info(
        f"Total Parameters        : {total_params:,}"
    )

    logger.info(
        f"Trainable Parameters    : {trainable_params:,}"
    )

    logger.info(
        f"Non Trainable Parameters: {non_trainable_params:,}"
    )

    logger.info("------------------------------------------")

    logger.info("=" * 70)

except Exception as e:

    logger.exception(
        f"Model architecture creation failed: {str(e)}"
    )

    raise e
```

**4.Freeze Feature Extractor**

**Implements**

```python


import logging
import tensorflow as tf


logger = logging.getLogger("RiceLeaf_EfficientNetB0")


try:

    logger.info("=" * 70)
    logger.info("Freezing EfficientNetB0 Feature Extractor")
    logger.info("=" * 70)


   
    # Freeze Backbone
   

    efficientnet_base.trainable = False


  
    # Count Layers
   

    total_layers = len(
        efficientnet_base.layers
    )


    trainable_layers = sum(

        1 for layer in efficientnet_base.layers

        if layer.trainable

    )


    frozen_layers = (
        total_layers - trainable_layers
    )


    
    # Parameter Statistics
   

    total_params = efficientnet_model.count_params()


    trainable_params = sum(

        tf.keras.backend.count_params(weight)

        for weight in efficientnet_model.trainable_weights

    )


    non_trainable_params = sum(

        tf.keras.backend.count_params(weight)

        for weight in efficientnet_model.non_trainable_weights

    )


    logger.info(
        "EfficientNetB0 backbone frozen successfully"
    )


    logger.info("------------------------------------------")


    logger.info(
        f"Total Backbone Layers    : {total_layers}"
    )


    logger.info(
        f"Trainable Backbone Layers: {trainable_layers}"
    )


    logger.info(
        f"Frozen Backbone Layers   : {frozen_layers}"
    )


    logger.info("------------------------------------------")


    logger.info(
        f"Total Parameters         : {total_params:,}"
    )


    logger.info(
        f"Trainable Parameters     : {trainable_params:,}"
    )


    logger.info(
        f"Non Trainable Parameters : {non_trainable_params:,}"
    )


    logger.info("------------------------------------------")


    logger.info("=" * 70)



except Exception as e:

    logger.exception(
        f"EfficientNetB0 freezing failed: {str(e)}"
    )

    raise e
```

**5.Compile Initial Model**


**Implements**


```python


import logging
import tensorflow as tf


logger = logging.getLogger("RiceLeaf_EfficientNetB0")


try:

    logger.info("=" * 70)
    logger.info("Compiling Initial EfficientNetB0 Transfer Learning Model")
    logger.info("=" * 70)


   
    # Optimizer Configuration
   

    optimizer = tf.keras.optimizers.Adam(

        learning_rate=
        EFFICIENTNET_CONFIG["INITIAL_LEARNING_RATE"]

    )


  
    # Compile Model
    

    efficientnet_model.compile(

        optimizer=optimizer,

        loss=tf.keras.losses.CategoricalCrossentropy(),

        metrics=[

            tf.keras.metrics.CategoricalAccuracy(
                name="accuracy"
            ),

            tf.keras.metrics.Precision(
                name="precision"
            ),

            tf.keras.metrics.Recall(
                name="recall"
            )

        ]

    )


    
    # Parameter Statistics
    

    trainable_params = sum(

        tf.keras.backend.count_params(weight)

        for weight in efficientnet_model.trainable_weights

    )


    non_trainable_params = sum(

        tf.keras.backend.count_params(weight)

        for weight in efficientnet_model.non_trainable_weights

    )


    logger.info(
        "EfficientNetB0 model compiled successfully"
    )


    logger.info("------------------------------------------")


    logger.info(
        "Optimizer              : Adam"
    )


    logger.info(
        f"Learning Rate          : "
        f"{EFFICIENTNET_CONFIG['INITIAL_LEARNING_RATE']}"
    )


    logger.info(
        "Loss                   : Categorical Crossentropy"
    )


    logger.info(
        f"Trainable Parameters   : {trainable_params:,}"
    )


    logger.info(
        f"Non Trainable Params   : {non_trainable_params:,}"
    )


    logger.info("------------------------------------------")

    logger.info("=" * 70)



except Exception as e:

    logger.exception(
        f"EfficientNetB0 compilation failed: {str(e)}"
    )

    raise e
```

**6.Configure Training Callbacks**

**Implements**

```python


import os
import logging
import tensorflow as tf


logger = logging.getLogger("RiceLeaf_EfficientNetB0")


try:

    logger.info("=" * 70)
    logger.info("Configuring EfficientNetB0 Training Callbacks")
    logger.info("=" * 70)


  
    # Create Directory Structure
   

    EFFICIENTNET_DIR = "efficientnetb0_training"

    os.makedirs(
        EFFICIENTNET_DIR,
        exist_ok=True
    )


    CHECKPOINT_PATH = os.path.join(
        EFFICIENTNET_DIR,
        "efficientnetb0_best.keras"
    )


    CSV_LOG_PATH = os.path.join(
        EFFICIENTNET_DIR,
        "efficientnetb0_training_log.csv"
    )


    TENSORBOARD_PATH = os.path.join(
        EFFICIENTNET_DIR,
        "tensorboard_logs"
    )


   
    # Callbacks
    

    EFFICIENTNET_CALLBACKS = [

        # Save best model
        tf.keras.callbacks.ModelCheckpoint(

            filepath=CHECKPOINT_PATH,

            monitor="val_accuracy",

            save_best_only=True,

            mode="max",

            verbose=1

        ),


        # Stop if no improvement
        tf.keras.callbacks.EarlyStopping(

            monitor="val_loss",

            patience=5,

            restore_best_weights=True,

            verbose=1

        ),


        # Reduce learning rate
        tf.keras.callbacks.ReduceLROnPlateau(

            monitor="val_loss",

            factor=0.2,

            patience=3,

            min_lr=1e-7,

            verbose=1

        ),


        # Training history logging
        tf.keras.callbacks.CSVLogger(

            CSV_LOG_PATH

        ),


        # TensorBoard
        tf.keras.callbacks.TensorBoard(

            log_dir=TENSORBOARD_PATH

        )

    ]


    logger.info(
        "EfficientNetB0 callbacks created successfully"
    )


    logger.info("------------------------------------------")


    logger.info(
        f"Checkpoint Path : {CHECKPOINT_PATH}"
    )


    logger.info(
        f"CSV Log Path   : {CSV_LOG_PATH}"
    )


    logger.info(
        f"TensorBoard    : {TENSORBOARD_PATH}"
    )


    logger.info(
        f"Total Callbacks: {len(EFFICIENTNET_CALLBACKS)}"
    )


    logger.info("------------------------------------------")


    logger.info("=" * 70)



except Exception as e:

    logger.exception(
        f"Callback configuration failed: {str(e)}"
    )

    raise e
```

**7.Dataset Path Configuration**

```python


import os
import logging


logger = logging.getLogger("RiceLeaf_Project")


try:

    logger.info("=" * 70)
    logger.info("Initializing Dataset Configuration")
    logger.info("=" * 70)


    
    # Detect Existing Dataset Structure
   

    DATASET_ROOT = "dataset"


    if not os.path.exists(DATASET_ROOT):

        raise FileNotFoundError(
            f"Dataset directory not found: {DATASET_ROOT}"
        )


    
    # Find Class Directory
   

    possible_paths = [

        os.path.join(DATASET_ROOT, "Data"),

        os.path.join(DATASET_ROOT, "data"),

        DATASET_ROOT

    ]


    DATA_DIR = None


    for path in possible_paths:

        if os.path.exists(path):

            DATA_DIR = path

            break



    if DATA_DIR is None:

        raise FileNotFoundError(
            "Could not locate dataset directory"
        )


    
    # Dataset Configuration
   

    DATASET_CONFIG = {

        "DATA_DIR": DATA_DIR,

        "IMAGE_SIZE": (224,224),

        "NUM_CLASSES": NUM_CLASSES

    }



    logger.info("------------------------------------------")

    logger.info(
        f"Dataset Location: {DATASET_CONFIG['DATA_DIR']}"
    )


    logger.info(
        f"Number of Classes: {DATASET_CONFIG['NUM_CLASSES']}"
    )


    logger.info("------------------------------------------")


    logger.info(
        "Dataset configuration initialized successfully"
    )


    logger.info("=" * 70)



except Exception as e:

    logger.exception(
        f"Dataset configuration failed: {str(e)}"
    )

    raise e
```

**8.EfficientNetB0 Data Generator**

**Implements:-**

```python


import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator
import logging


logger = logging.getLogger("RiceLeaf_EfficientNetB0")


try:


    logger.info("="*70)
    logger.info("Creating EfficientNetB0 Data Pipeline")
    logger.info("="*70)



    
    # Training Data Generator
   

    efficientnet_train_datagen = ImageDataGenerator(

        validation_split=0.2,


        preprocessing_function=
        tf.keras.applications.efficientnet.preprocess_input,


        # Moderate augmentation
        rotation_range=15,

        width_shift_range=0.10,

        height_shift_range=0.10,

        zoom_range=0.15,

        shear_range=0.10,

        horizontal_flip=True,


        fill_mode="nearest"

    )



    
    # Validation Generator
    

    efficientnet_validation_datagen = ImageDataGenerator(

        validation_split=0.2,


        preprocessing_function=
        tf.keras.applications.efficientnet.preprocess_input

    )



   
    # Training Generator
    

    efficientnet_train_generator = (
        
        efficientnet_train_datagen.flow_from_directory(

            DATASET_CONFIG["DATA_DIR"],


            target_size=
            EFFICIENTNET_CONFIG["IMG_SIZE"],


            batch_size=
            EFFICIENTNET_CONFIG["BATCH_SIZE"],


            class_mode="categorical",


            subset="training",


            shuffle=True,


            seed=SEED

        )
    )



   
    # Validation Generator
    

    efficientnet_validation_generator = (

        efficientnet_validation_datagen.flow_from_directory(

            DATASET_CONFIG["DATA_DIR"],


            target_size=
            EFFICIENTNET_CONFIG["IMG_SIZE"],


            batch_size=
            EFFICIENTNET_CONFIG["BATCH_SIZE"],


            class_mode="categorical",


            subset="validation",


            shuffle=False,


            seed=SEED

        )
    )



    
    # Dataset Information
    

    logger.info("------------------------------------------")


    logger.info(
        f"Training Samples : "
        f"{efficientnet_train_generator.samples}"
    )


    logger.info(
        f"Validation Samples : "
        f"{efficientnet_validation_generator.samples}"
    )


    logger.info(
        f"Class Mapping : "
        f"{efficientnet_train_generator.class_indices}"
    )


    logger.info(
        f"Image Size : "
        f"{EFFICIENTNET_CONFIG['IMG_SIZE']}"
    )


    logger.info("------------------------------------------")



except Exception as e:


    logger.exception(
        f"EfficientNet data generator failed: {str(e)}"
    )

    raise e
```

**9.Train Classification Head**

**Implements:-**

```python


import logging
import time


logger = logging.getLogger("RiceLeaf_EfficientNetB0")


try:

    logger.info("=" * 70)
    logger.info("Starting EfficientNetB0 Classification Head Training")
    logger.info("=" * 70)


    
    # Training Start Time
    

    start_time = time.time()



    # Model Training
   

    efficientnet_history = efficientnet_model.fit(

        efficientnet_train_generator,

        validation_data=
        efficientnet_validation_generator,

        epochs=
        EFFICIENTNET_CONFIG["INITIAL_EPOCHS"],

        callbacks=
        EFFICIENTNET_CALLBACKS,

        verbose=1

    )


    # Training Duration


    end_time = time.time()


    training_time = (

        end_time - start_time

    ) / 60



    # Best Validation Accuracy
    

    best_val_accuracy = max(

        efficientnet_history.history[
            "val_accuracy"
        ]

    )


    best_epoch = (

        efficientnet_history.history[
            "val_accuracy"
        ].index(best_val_accuracy)

        + 1

    )



    # Logging Results
  

    logger.info("------------------------------------------")


    logger.info(
        "Classification head training completed successfully"
    )


    logger.info("------------------------------------------")


    logger.info(
        f"Training Epochs          : "
        f"{len(efficientnet_history.history['loss'])}"
    )


    logger.info(
        f"Training Time            : "
        f"{training_time:.2f} minutes"
    )


    logger.info(
        f"Best Epoch               : "
        f"{best_epoch}"
    )


    logger.info(
        f"Best Validation Accuracy : "
        f"{best_val_accuracy:.4f}"
    )


    logger.info("------------------------------------------")


    logger.info("=" * 70)



except Exception as e:


    logger.exception(

        f"EfficientNetB0 training failed: {str(e)}"

    )


    raise e
```

**10.Evaluate Initial Model**

**Implements:-**

```python


import logging
import numpy as np

from sklearn.metrics import (
    classification_report,
    precision_score,
    recall_score,
    f1_score
)


logger = logging.getLogger("RiceLeaf_EfficientNetB0")


try:

    logger.info("=" * 70)
    logger.info("Starting Initial EfficientNetB0 Model Evaluation")
    logger.info("=" * 70)


   
    # Model Evaluation
   

    evaluation_results = efficientnet_model.evaluate(

        efficientnet_validation_generator,

        verbose=1

    )


    
    # Predictions
    

    predictions = efficientnet_model.predict(

        efficientnet_validation_generator,

        verbose=1

    )


    y_pred = np.argmax(
        predictions,
        axis=1
    )


    y_true = efficientnet_validation_generator.classes



  
    # Metrics
   

    precision = precision_score(

        y_true,

        y_pred,

        average="weighted"

    )


    recall = recall_score(

        y_true,

        y_pred,

        average="weighted"

    )


    f1 = f1_score(

        y_true,

        y_pred,

        average="weighted"

    )


    accuracy = np.mean(

        y_true == y_pred

    )


   
    # Logging
    

    logger.info("------------------------------------------")


    logger.info(
        f"Loss      : {evaluation_results[0]:.4f}"
    )


    logger.info(
        f"Accuracy  : {accuracy:.4f}"
    )


    logger.info(
        f"Precision : {precision:.4f}"
    )


    logger.info(
        f"Recall    : {recall:.4f}"
    )


    logger.info(
        f"F1 Score  : {f1:.4f}"
    )


    logger.info("------------------------------------------")


    print("\nInitial EfficientNetB0 Classification Report\n")


    print(

        classification_report(

            y_true,

            y_pred,

            target_names=
            list(
                efficientnet_validation_generator.class_indices.keys()
            )

        )

    )


    logger.info(
        "Initial EfficientNetB0 evaluation completed successfully"
    )


    logger.info("=" * 70)



except Exception as e:


    logger.exception(

        f"EfficientNetB0 evaluation failed: {str(e)}"

    )


    raise e
```

**11.Fine-Tune EfficientNetB0**

**Implements:-**

```python

import logging
import tensorflow as tf


logger = logging.getLogger("RiceLeaf_EfficientNetB0")


try:

    logger.info("=" * 70)
    logger.info("Starting EfficientNetB0 Fine-Tuning Configuration")
    logger.info("=" * 70)


    
    # Unfreeze EfficientNet Backbone
    

    efficientnet_base.trainable = True


    total_layers = len(
        efficientnet_base.layers
    )


    # Calculate layers to unfreeze

    unfreeze_layers = int(

        total_layers *
        EFFICIENTNET_CONFIG["UNFREEZE_PERCENTAGE"]

    )


    freeze_until = (

        total_layers -
        unfreeze_layers

    )


    
    # Freeze Early Layers
   

    for layer in efficientnet_base.layers[:freeze_until]:

        layer.trainable = False



    
    # Unfreeze Last Layers
    

    for layer in efficientnet_base.layers[freeze_until:]:

        layer.trainable = True



   
    # Parameter Statistics
    

    trainable_params = sum(

        tf.keras.backend.count_params(weight)

        for weight in efficientnet_model.trainable_weights

    )


    non_trainable_params = sum(

        tf.keras.backend.count_params(weight)

        for weight in efficientnet_model.non_trainable_weights

    )


    trainable_layers = sum(

        1 for layer in efficientnet_base.layers

        if layer.trainable

    )


    frozen_layers = (

        total_layers -
        trainable_layers

    )


    logger.info("------------------------------------------")


    logger.info(
        "EfficientNetB0 fine-tuning configured successfully"
    )


    logger.info("------------------------------------------")


    logger.info(
        f"Total Backbone Layers : {total_layers}"
    )


    logger.info(
        f"Trainable Layers      : {trainable_layers}"
    )


    logger.info(
        f"Frozen Layers         : {frozen_layers}"
    )


    logger.info("------------------------------------------")


    logger.info(
        f"Trainable Parameters  : {trainable_params:,}"
    )


    logger.info(
        f"Non Trainable Params  : {non_trainable_params:,}"
    )


    logger.info("------------------------------------------")


    logger.info("=" * 70)



except Exception as e:


    logger.exception(

        f"EfficientNetB0 fine-tuning setup failed: {str(e)}"

    )

    raise e
```

**13. Train Fine-Tuned Model**

**14. Performance Analysis**

**Implements:-**

```python



import logging
import time


logger = logging.getLogger(
    "RiceLeaf_EfficientNetB0"
)



try:


    logger.info("="*70)

    logger.info(
        "Starting EfficientNetB0 Performance Analysis"
    )

    logger.info("="*70)



   
    # Validate History
    


    history = efficientnet_finetune_history.history



    required_metrics = [

        "accuracy",
        "val_accuracy",
        "loss",
        "val_loss"

    ]


    for metric in required_metrics:

        if metric not in history:

            raise ValueError(
                f"Missing metric: {metric}"
            )



   
    # Extract Metrics
    


    train_accuracy = history["accuracy"]

    val_accuracy = history["val_accuracy"]

    train_loss = history["loss"]

    val_loss = history["val_loss"]



    
    # Best Performance
    


    best_epoch = (

        val_accuracy.index(
            max(val_accuracy)
        )

        + 1

    )


    best_val_accuracy = max(
        val_accuracy
    )


    best_val_loss = min(
        val_loss
    )



    
    # Final Performance
    


    final_train_accuracy = train_accuracy[-1]

    final_val_accuracy = val_accuracy[-1]


    final_train_loss = train_loss[-1]

    final_val_loss = val_loss[-1]



    
    # Generalization Analysis
   


    accuracy_gap = (

        final_train_accuracy -
        final_val_accuracy

    )


    loss_gap = (

        final_val_loss -
        final_train_loss

    )



    
    # Parameter Information
    


    total_parameters = (

        efficientnet_model.count_params()

    )



   
    # Logging
 


    logger.info("------------------------------------------")


    logger.info(
        f"Best Epoch              : {best_epoch}"
    )


    logger.info(
        f"Best Validation Accuracy: {best_val_accuracy:.4f}"
    )


    logger.info(
        f"Best Validation Loss    : {best_val_loss:.4f}"
    )


    logger.info("------------------------------------------")


    logger.info(
        f"Final Train Accuracy    : {final_train_accuracy:.4f}"
    )


    logger.info(
        f"Final Validation Accuracy: {final_val_accuracy:.4f}"
    )


    logger.info("------------------------------------------")


    logger.info(
        f"Accuracy Gap            : {accuracy_gap:.4f}"
    )


    logger.info(
        f"Loss Gap                : {loss_gap:.4f}"
    )


    logger.info("------------------------------------------")


    logger.info(
        f"Total Parameters        : {total_parameters:,}"
    )



  
    # Overfitting Detection
    


    if accuracy_gap > 0.15 or loss_gap > 0.5:


        logger.warning(
            "Model shows possible overfitting"
        )


    else:


        logger.info(
            "Model generalization looks acceptable"
        )



    logger.info("="*70)

    logger.info(
        "Performance Analysis Completed"
    )

    logger.info("="*70)



except Exception as e:


    logger.exception(

        f"Performance analysis failed: {str(e)}"

    )

    raise e
```

```python



import logging
import numpy as np


from sklearn.metrics import (

    classification_report,
    confusion_matrix,
    precision_score,
    recall_score,
    f1_score,
    accuracy_score,
    roc_auc_score

)


logger = logging.getLogger(
    "RiceLeaf_EfficientNetB0"
)



try:


    logger.info("="*70)

    logger.info(
        "Starting Final EfficientNetB0 Model Evaluation"
    )

    logger.info("="*70)



    
    # Model Evaluation
   


    evaluation_results = efficientnet_model.evaluate(

        efficientnet_validation_generator,

        verbose=1

    )



    
    # Predictions
   


    predictions = efficientnet_model.predict(

        efficientnet_validation_generator,

        verbose=1

    )



    y_pred = np.argmax(

        predictions,

        axis=1

    )



    y_true = (

        efficientnet_validation_generator.classes

    )



    if len(y_true) != len(y_pred):

        raise ValueError(
            "Prediction and label size mismatch"
        )



    class_names = list(

        efficientnet_validation_generator.class_indices.keys()

    )



    
    # Metrics
    


    accuracy = accuracy_score(

        y_true,

        y_pred

    )


    precision_weighted = precision_score(

        y_true,

        y_pred,

        average="weighted",

        zero_division=0

    )


    recall_weighted = recall_score(

        y_true,

        y_pred,

        average="weighted",

        zero_division=0

    )


    f1_weighted = f1_score(

        y_true,

        y_pred,

        average="weighted",

        zero_division=0

    )


    f1_macro = f1_score(

        y_true,

        y_pred,

        average="macro",

        zero_division=0

    )



    
    # Logging
    


    logger.info("------------------------------------------")


    logger.info(
        f"Loss               : {evaluation_results[0]:.4f}"
    )


    logger.info(
        f"Accuracy           : {accuracy:.4f}"
    )


    logger.info(
        f"Precision Weighted : {precision_weighted:.4f}"
    )


    logger.info(
        f"Recall Weighted    : {recall_weighted:.4f}"
    )


    logger.info(
        f"F1 Weighted        : {f1_weighted:.4f}"
    )


    logger.info(
        f"F1 Macro           : {f1_macro:.4f}"
    )


    logger.info("------------------------------------------")



   
    # Reports
    


    print(
        classification_report(

            y_true,

            y_pred,

            target_names=class_names,

            zero_division=0

        )
    )



    cm = confusion_matrix(

        y_true,

        y_pred

    )


    print("\nConfusion Matrix\n")

    print(cm)



    logger.info(
        "Final Evaluation Completed Successfully"
    )


    logger.info("="*70)



except Exception as e:


    logger.exception(

        f"Evaluation failed: {str(e)}"

    )


    raise e
```

```python
# ============================================================
# 13.2 TRAINING HISTORY EXTRACTION - MOBILENETV2
# ============================================================


import logging
import pandas as pd


logger = logging.getLogger(
    "RiceLeaf_MobileNetV2"
)



try:


    logger.info("="*70)

    logger.info(
        "Extracting MobileNetV2 Training History"
    )

    logger.info("="*70)



    # --------------------------------------------------------
    # Extract History Object
    # --------------------------------------------------------


    history = (
        mobilenet_finetune_history.history
    )



    # --------------------------------------------------------
    # Validate Required Metrics
    # --------------------------------------------------------


    required_metrics = [

        "accuracy",

        "loss",

        "val_accuracy",

        "val_loss"

    ]



    missing_metrics = [

        metric

        for metric in required_metrics

        if metric not in history

    ]



    if missing_metrics:

        raise ValueError(

            f"Missing training metrics: {missing_metrics}"

        )



    # --------------------------------------------------------
    # Convert History To DataFrame
    # --------------------------------------------------------


    history_df = pd.DataFrame(

        history

    )



    # Add Epoch Column


    history_df.insert(

        0,

        "epoch",

        range(

            1,

            len(history_df)+1

        )

    )



    # --------------------------------------------------------
    # Training Statistics
    # --------------------------------------------------------


    total_epochs = len(

        history_df

    )


    logger.info("------------------------------------------")


    logger.info(

        f"Total Training Epochs: {total_epochs}"

    )


    logger.info(

        f"Available Metrics: {list(history_df.columns)}"

    )


    logger.info("------------------------------------------")



    # --------------------------------------------------------
    # Display Preview
    # --------------------------------------------------------


    print("\nTraining History Preview\n")

    display(

        history_df.head()

    )



    logger.info(

        "Training history extraction completed successfully"

    )


    logger.info("="*70)



except Exception as e:


    logger.exception(

        f"Training history extraction failed: {str(e)}"

    )


    raise e
```

**6.Learning Behavior & Convergence Analysis**

**Implements:-**

 * This step analyzes how the model learned over epochs.

  * Accuracy improvement rate
  * Loss reduction rate
*  Training stability
  * Whether the model has converged
  * Whether more epochs are useful

```python



import logging
import numpy as np



logger = logging.getLogger(
    "RiceLeaf_MobileNetV2"
)



try:


    logger.info("="*70)

    logger.info(
        "Analyzing MobileNetV2 Learning Behavior and Convergence"
    )

    logger.info("="*70)



    
    # Accuracy Improvement Analysis
 


    initial_accuracy = (

        training_analysis_df["val_accuracy"].iloc[0]

    )


    final_accuracy = (

        training_analysis_df["val_accuracy"].iloc[-1]

    )


    accuracy_improvement = (

        final_accuracy -

        initial_accuracy

    )



    
    # Loss Reduction Analysis
    


    initial_loss = (

        training_analysis_df["val_loss"].iloc[0]

    )


    final_loss = (

        training_analysis_df["val_loss"].iloc[-1]

    )


    loss_reduction = (

        initial_loss -

        final_loss

    )



    
    # Epoch Stability Analysis
 


    accuracy_changes = (

        training_analysis_df["val_accuracy"]

        .diff()

        .abs()

    )


    average_accuracy_change = (

        accuracy_changes.mean()

    )



    
    # Convergence Detection
    


    convergence_threshold = 0.01



    if average_accuracy_change < convergence_threshold:


        convergence_status = (

            "Model has converged"

        )


    else:


        convergence_status = (

            "Model is still learning"

        )



    
    # Logging Results
    


    logger.info("------------------------------------------")


    logger.info(

        f"Initial Validation Accuracy : {initial_accuracy:.4f}"

    )


    logger.info(

        f"Final Validation Accuracy   : {final_accuracy:.4f}"

    )


    logger.info(

        f"Accuracy Improvement        : {accuracy_improvement:.4f}"

    )


    logger.info("------------------------------------------")


    logger.info(

        f"Initial Validation Loss     : {initial_loss:.4f}"

    )


    logger.info(

        f"Final Validation Loss       : {final_loss:.4f}"

    )


    logger.info(

        f"Loss Reduction              : {loss_reduction:.4f}"

    )


    logger.info("------------------------------------------")


    logger.info(

        f"Average Accuracy Change     : {average_accuracy_change:.4f}"

    )


    logger.info(

        f"Convergence Status          : {convergence_status}"

    )


    logger.info("------------------------------------------")


    logger.info(

        "Learning behavior analysis completed successfully"

    )


    logger.info("="*70)



except Exception as e:


    logger.exception(

        f"Learning behavior analysis failed: {str(e)}"

    )


    raise e
```

**7.Overfitting and Underfitting Diagnosis**

**Implements:-**

```python


import logging



logger = logging.getLogger(
    "RiceLeaf_MobileNetV2"
)



try:


    logger.info("="*70)

    logger.info(
        "Running Overfitting and Underfitting Detection"
    )

    logger.info("="*70)



    
    # Extract Final Metrics
   

    final_train_accuracy = (

        training_analysis_df["accuracy"].iloc[-1]

    )


    final_val_accuracy = (

        training_analysis_df["val_accuracy"].iloc[-1]

    )


    final_train_loss = (

        training_analysis_df["loss"].iloc[-1]

    )


    final_val_loss = (

        training_analysis_df["val_loss"].iloc[-1]

    )



   
    # Calculate Gaps
   


    accuracy_gap = (

        final_train_accuracy

        -

        final_val_accuracy

    )


    loss_gap = (

        final_val_loss

        -

        final_train_loss

    )



   
    # Detection Thresholds
   


    OVERFIT_ACCURACY_THRESHOLD = 0.10

    UNDERFIT_ACCURACY_THRESHOLD = 0.70



    
    # Model Behavior Classification
    


    if (

        accuracy_gap > OVERFIT_ACCURACY_THRESHOLD

        and

        loss_gap > 0

    ):


        model_status = (

            "Overfitting Detected"

        )



        recommendation = (

            "Increase regularization, augmentation, or reduce model complexity"

        )



    elif (

        final_train_accuracy < UNDERFIT_ACCURACY_THRESHOLD

        and

        final_val_accuracy < UNDERFIT_ACCURACY_THRESHOLD

    ):


        model_status = (

            "Underfitting Detected"

        )


        recommendation = (

            "Increase training duration or improve feature learning"

        )



    else:


        model_status = (

            "Good Generalization"

        )


        recommendation = (

            "Model training behavior is acceptable"

        )



  
    # Logging Results
    


    logger.info("------------------------------------------")


    logger.info(

        f"Final Training Accuracy   : {final_train_accuracy:.4f}"

    )


    logger.info(

        f"Final Validation Accuracy : {final_val_accuracy:.4f}"

    )


    logger.info(

        f"Accuracy Gap              : {accuracy_gap:.4f}"

    )


    logger.info("------------------------------------------")


    logger.info(

        f"Final Training Loss       : {final_train_loss:.4f}"

    )


    logger.info(

        f"Final Validation Loss     : {final_val_loss:.4f}"

    )


    logger.info(

        f"Loss Gap                  : {loss_gap:.4f}"

    )


    logger.info("------------------------------------------")


    logger.info(

        f"Model Status              : {model_status}"

    )


    logger.info(

        f"Recommendation             : {recommendation}"

    )


    logger.info("------------------------------------------")


    logger.info(

        "Overfitting/Underfitting analysis completed successfully"

    )


    logger.info("="*70)



except Exception as e:


    logger.exception(

        f"Overfitting analysis failed: {str(e)}"

    )


    raise e
```

**10. Learning Rate Schedule Analysis**

**Implements:-**

**1.Learning rate analysis helps understand:**

* How optimizer adjusted learning rate
* Whether training stabilized
* Whether ReduceLROnPlateau was triggered
* Final learning rate behavior




```python



import logging
import pandas as pd
import os
import matplotlib.pyplot as plt



logger = logging.getLogger(
    "RiceLeaf_MobileNetV2"
)



try:


    logger.info("="*70)

    logger.info(
        "Analyzing MobileNetV2 Learning Rate Schedule"
    )

    logger.info("="*70)



    
    # Extract Learning Rate History
    


    history_df = training_analysis_df.copy()



    if "learning_rate" not in history_df.columns:


        raise ValueError(
            "Learning rate information not available in training history"
        )



    learning_rates = history_df["learning_rate"]



    epochs = range(

        1,

        len(learning_rates) + 1

    )



    
    # Learning Rate Statistics
    


    initial_lr = learning_rates.iloc[0]


    final_lr = learning_rates.iloc[-1]


    minimum_lr = learning_rates.min()



    lr_changed = (

        initial_lr != final_lr

    )



    lr_reduction_count = (

        sum(

            learning_rates.diff() < 0

        )

    )



    
    # Logging Analysis
    


    logger.info("------------------------------------------")


    logger.info(

        f"Initial Learning Rate : {initial_lr:.8f}"

    )


    logger.info(

        f"Final Learning Rate   : {final_lr:.8f}"

    )


    logger.info(

        f"Minimum Learning Rate : {minimum_lr:.8f}"

    )


    logger.info(

        f"Learning Rate Reduced: {lr_changed}"

    )


    logger.info(

        f"LR Reduction Events   : {lr_reduction_count}"

    )


    logger.info("------------------------------------------")



   
    # Plot Learning Rate Curve
   


    LR_DIR = os.path.join(

        "artifacts",

        "training_curves"

    )


    os.makedirs(

        LR_DIR,

        exist_ok=True

    )


    lr_plot_path = os.path.join(

        LR_DIR,

        "mobilenetv2_learning_rate_curve.png"

    )



    plt.figure(

        figsize=(8,5)

    )


    plt.plot(

        epochs,

        learning_rates,

        marker="o"

    )


    plt.xlabel(

        "Epoch"

    )


    plt.ylabel(

        "Learning Rate"

    )


    plt.title(

        "MobileNetV2 Learning Rate Schedule"

    )


    plt.grid(True)



    plt.savefig(

        lr_plot_path,

        bbox_inches="tight"

    )


    plt.close()



    logger.info(

        f"Learning Rate Curve Saved: {lr_plot_path}"

    )



    
    # Save Learning Rate Data
    


    lr_csv_path = os.path.join(

        LR_DIR,

        "mobilenetv2_learning_rate_history.csv"

    )


    history_df[

        [

            "epoch",

            "learning_rate"

        ]

    ].to_csv(

        lr_csv_path,

        index=False

    )



    logger.info(

        f"Learning Rate History Saved: {lr_csv_path}"

    )



    logger.info("------------------------------------------")



    if lr_changed:


        logger.info(

            "Learning rate scheduler adjusted the training process"

        )


    else:


        logger.info(

            "Learning rate remained stable during training"

        )



    logger.info("------------------------------------------")


    logger.info(

        "Learning Rate Schedule Analysis Completed Successfully"

    )


    logger.info("="*70)



except Exception as e:


    logger.exception(

        f"Learning rate analysis failed: {str(e)}"

    )


    raise e
```

```python



import logging
import os
import json



logger = logging.getLogger(
    "RiceLeaf_MobileNetV2"
)



try:


    logger.info("="*70)

    logger.info(
        "Running MobileNetV2 Generalization Performance Assessment"
    )

    logger.info("="*70)



   
    # Extract Final Metrics
    


    final_train_accuracy = (

        training_analysis_df["accuracy"].iloc[-1]

    )


    final_val_accuracy = (

        training_analysis_df["val_accuracy"].iloc[-1]

    )


    best_val_accuracy = (

        training_analysis_df["val_accuracy"].max()

    )


    final_train_loss = (

        training_analysis_df["loss"].iloc[-1]

    )


    final_val_loss = (

        training_analysis_df["val_loss"].iloc[-1]

    )



    
    # Generalization Gap Calculation
    


    accuracy_gap = (

        final_train_accuracy -

        final_val_accuracy

    )


    loss_gap = (

        final_train_loss -

        final_val_loss

    )



   
    # Model Behavior Classification
    


    if accuracy_gap > 0.15:


        model_status = (

            "Overfitting Detected"

        )


        recommendation = (

            "Apply stronger regularization"

        )



    elif final_train_accuracy < 0.75 and final_val_accuracy < 0.75:


        model_status = (

            "Underfitting Detected"

        )


        recommendation = (

            "Increase training complexity"

        )



    else:


        model_status = (

            "Good Generalization"

        )


        recommendation = (

            "Model is suitable for deployment"

        )



   
    # Create Assessment Report
   


    generalization_report = {


        "Model":

        "MobileNetV2",



        "Final_Training_Accuracy":

        round(float(final_train_accuracy),4),



        "Final_Validation_Accuracy":

        round(float(final_val_accuracy),4),



        "Best_Validation_Accuracy":

        round(float(best_val_accuracy),4),



        "Accuracy_Gap":

        round(float(accuracy_gap),4),



        "Final_Training_Loss":

        round(float(final_train_loss),4),



        "Final_Validation_Loss":

        round(float(final_val_loss),4),



        "Loss_Gap":

        round(float(loss_gap),4),



        "Model_Status":

        model_status,



        "Recommendation":

        recommendation

    }



    
    # Logging
   


    logger.info("------------------------------------------")


    logger.info(

        f"Final Training Accuracy   : {final_train_accuracy:.4f}"

    )


    logger.info(

        f"Final Validation Accuracy : {final_val_accuracy:.4f}"

    )


    logger.info(

        f"Best Validation Accuracy  : {best_val_accuracy:.4f}"

    )


    logger.info(

        f"Accuracy Gap              : {accuracy_gap:.4f}"

    )


    logger.info("------------------------------------------")


    logger.info(

        f"Final Training Loss       : {final_train_loss:.4f}"

    )


    logger.info(

        f"Final Validation Loss     : {final_val_loss:.4f}"

    )


    logger.info(

        f"Loss Gap                  : {loss_gap:.4f}"

    )


    logger.info("------------------------------------------")


    logger.info(

        f"Model Status              : {model_status}"

    )


    logger.info(

        f"Recommendation            : {recommendation}"

    )


    logger.info("------------------------------------------")



    
    # Save Report
   


    REPORT_DIR = os.path.join(

        "artifacts",

        "training_summary"

    )


    os.makedirs(

        REPORT_DIR,

        exist_ok=True

    )



    report_path = os.path.join(

        REPORT_DIR,

        "mobilenetv2_generalization_report.json"

    )



    with open(

        report_path,

        "w"

    ) as file:


        json.dump(

            generalization_report,

            file,

            indent=4

        )



    logger.info(

        f"Generalization Report Saved: {report_path}"

    )



    logger.info(

        "Generalization Performance Assessment Completed Successfully"

    )


    logger.info("="*70)



except Exception as e:


    logger.exception(

        f"Generalization assessment failed: {str(e)}"

    )


    raise e
```

**2.Validation Dataset Preparation**

**Implements:-**

```python



import logging
import numpy as np



logger = logging.getLogger(
    "RiceLeaf_MobileNetV2_Prediction"
)



try:


    logger.info("="*70)

    logger.info(
        "Preparing Validation Dataset for MobileNetV2 Inference"
    )

    logger.info("="*70)



    # --------------------------------------------------------
    # Validate Generator
    # --------------------------------------------------------


    if "mobilenet_validation_generator" not in globals():

        raise ValueError(
            "MobileNetV2 validation generator not found"
        )



    logger.info(
        "Validation generator verified"
    )



    
    # Reset Generator
   


    mobilenet_validation_generator.reset()



    logger.info(
        "Validation generator reset completed"
    )



    
    # Extract Dataset Information
    


    validation_samples = (

        mobilenet_validation_generator.samples

    )


    class_indices = (

        mobilenet_validation_generator.class_indices

    )


    class_names = list(

        class_indices.keys()

    )



   
    # Extract True Labels
   


    y_true = (

        mobilenet_validation_generator.classes

    )



  
    # Verify Class Distribution
   


    unique_classes, class_counts = np.unique(

        y_true,

        return_counts=True

    )


    class_distribution = dict(

        zip(

            [
                class_names[i]
                for i in unique_classes
            ],

            class_counts

        )

    )



    
    # Get Batch Information
   


    sample_batch, sample_labels = next(

        mobilenet_validation_generator

    )


    batch_image_shape = (

        sample_batch.shape

    )


    batch_label_shape = (

        sample_labels.shape

    )



    # Reset again after sampling

    mobilenet_validation_generator.reset()



    
    # Logging
    


    logger.info("------------------------------------------")


    logger.info(

        f"Validation Samples       : {validation_samples}"

    )


    logger.info(

        f"Image Batch Shape        : {batch_image_shape}"

    )


    logger.info(

        f"Label Batch Shape        : {batch_label_shape}"

    )


    logger.info(

        f"Class Mapping            : {class_indices}"

    )


    logger.info(

        f"Class Distribution       : {class_distribution}"

    )


    logger.info("------------------------------------------")



   
   


    INFERENCE_METADATA = {


        "validation_samples":

        int(validation_samples),


        "class_names":

        class_names,


        "class_indices":

        class_indices,


        "class_distribution":

        class_distribution,


        "image_shape":

        str(batch_image_shape),


        "label_shape":

        str(batch_label_shape)

    }



    logger.info(

        "Validation dataset preparation completed successfully"

    )


    logger.info("="*70)



except Exception as e:


    logger.exception(

        f"Validation dataset preparation failed: {str(e)}"

    )


    raise e
```

**3.Generate Model Predictions**

**Implements:-**

```python



import logging
import numpy as np



logger = logging.getLogger(
    "RiceLeaf_MobileNetV2_Prediction"
)



try:


    logger.info("="*70)

    logger.info(
        "Generating MobileNetV2 Predictions"
    )

    logger.info("="*70)



    
    # Validate Required Objects
    


    required_objects = [

        "mobilenet_model",

        "mobilenet_validation_generator",

        "class_names"

    ]



    for obj in required_objects:


        if obj not in globals():

            raise ValueError(
                f"{obj} is not available"
            )



    logger.info(
        "Required prediction objects verified"
    )



    
    # Reset Generator Before Prediction
   


    mobilenet_validation_generator.reset()



    logger.info(
        "Validation generator reset"
    )



   
    # Generate Predictions
   


    prediction_probabilities = (

        mobilenet_model.predict(

            mobilenet_validation_generator,

            verbose=1

        )

    )



    logger.info(
        "Model prediction completed"
    )



   
    # Convert Probabilities to Classes
    


    y_pred = np.argmax(

        prediction_probabilities,

        axis=1

    )



    y_true = (

        mobilenet_validation_generator.classes

    )



  
    # Extract Confidence Scores
   


    confidence_scores = np.max(

        prediction_probabilities,

        axis=1

    )



   
    # Convert Class Index to Label
    


    prediction_labels = [

        class_names[index]

        for index in y_pred

    ]



    true_labels = [

        class_names[index]

        for index in y_true

    ]



  
    # Prediction Statistics
    

    total_predictions = len(

        prediction_labels

    )


    average_confidence = np.mean(

        confidence_scores

    )



    maximum_confidence = np.max(

        confidence_scores

    )



    minimum_confidence = np.min(

        confidence_scores

    )



   
    # Logging Results
  


    logger.info("------------------------------------------")


    logger.info(

        f"Total Predictions      : {total_predictions}"

    )


    logger.info(

        f"Prediction Shape       : {prediction_probabilities.shape}"

    )


    logger.info(

        f"Average Confidence     : {average_confidence:.4f}"

    )


    logger.info(

        f"Maximum Confidence     : {maximum_confidence:.4f}"

    )


    logger.info(

        f"Minimum Confidence     : {minimum_confidence:.4f}"

    )


    logger.info("------------------------------------------")



    logger.info(

        "Prediction generation completed successfully"

    )


    logger.info("="*70)



except Exception as e:


    logger.exception(

        f"Prediction generation failed: {str(e)}"

    )


    raise e
```

**4.Prediction Confidence Analysis**

```python



import logging
import numpy as np



logger = logging.getLogger(
    "RiceLeaf_MobileNetV2_Prediction"
)



try:


    logger.info("="*70)

    logger.info(
        "Running MobileNetV2 Prediction Confidence Analysis"
    )

    logger.info("="*70)



    
    # Validate Confidence Data
   


    if "confidence_scores" not in globals():

        raise ValueError(
            "Confidence scores not available"
        )



   
    # Convert Confidence to Percentage
   


    confidence_percentage = (

        confidence_scores * 100

    )



    
    # Statistical Analysis
  


    average_confidence = np.mean(

        confidence_percentage

    )


    maximum_confidence = np.max(

        confidence_percentage

    )


    minimum_confidence = np.min(

        confidence_percentage

    )



    max_confidence_index = np.argmax(

        confidence_scores

    )


    min_confidence_index = np.argmin(

        confidence_scores

    )



   
    # Confidence Categorization
    


    high_confidence_count = np.sum(

        confidence_percentage >= 90

    )


    medium_confidence_count = np.sum(

        (

            confidence_percentage >= 70

        )

        &

        (

            confidence_percentage < 90

        )

    )


    low_confidence_count = np.sum(

        confidence_percentage < 70

    )



    total_predictions = len(

        confidence_percentage

    )



    
    # Confidence Distribution
   


    confidence_distribution = {


        "High Confidence (>=90%)":

        int(high_confidence_count),



        "Medium Confidence (70-90%)":

        int(medium_confidence_count),



        "Low Confidence (<70%)":

        int(low_confidence_count)

    }



  
    # Logging Results
  


    logger.info("------------------------------------------")


    logger.info(

        f"Total Predictions        : {total_predictions}"

    )


    logger.info(

        f"Average Confidence       : {average_confidence:.2f}%"

    )


    logger.info(

        f"Maximum Confidence       : {maximum_confidence:.2f}%"

    )


    logger.info(

        f"Minimum Confidence       : {minimum_confidence:.2f}%"

    )


    logger.info("------------------------------------------")


    logger.info(

        f"Highest Confidence Image : Index {max_confidence_index}"

    )


    logger.info(

        f"Lowest Confidence Image  : Index {min_confidence_index}"

    )


    logger.info("------------------------------------------")


    logger.info(

        f"High Confidence Samples  : {high_confidence_count}"

    )


    logger.info(

        f"Medium Confidence Samples: {medium_confidence_count}"

    )


    logger.info(

        f"Low Confidence Samples   : {low_confidence_count}"

    )


    logger.info("------------------------------------------")


    logger.info(

        f"Confidence Distribution : {confidence_distribution}"

    )



    logger.info(

        "Prediction confidence analysis completed successfully"

    )


    logger.info("="*70)



except Exception as e:


    logger.exception(

        f"Confidence analysis failed: {str(e)}"

    )


    raise e
```

**5.Prediction Results Summary**

**Implements:-**

```python


import logging
import numpy as np

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)



logger = logging.getLogger(
    "RiceLeaf_MobileNetV2_Prediction"
)



try:


    logger.info("="*70)

    logger.info(
        "Creating MobileNetV2 Prediction Results Summary"
    )

    logger.info("="*70)



   
    # Validate Prediction Variables
    


    required_variables = [

        "y_true",

        "y_pred"

    ]


    for variable in required_variables:


        if variable not in globals():

            raise ValueError(
                f"{variable} not available"
            )



   
    # Calculate Performance Metrics
   


    prediction_accuracy = accuracy_score(

        y_true,

        y_pred

    )



    prediction_precision = precision_score(

        y_true,

        y_pred,

        average="weighted",

        zero_division=0

    )



    prediction_recall = recall_score(

        y_true,

        y_pred,

        average="weighted",

        zero_division=0

    )



    prediction_f1 = f1_score(

        y_true,

        y_pred,

        average="weighted",

        zero_division=0

    )



   
    # Prediction Statistics
   


    total_predictions = len(

        y_true

    )



    correct_predictions = np.sum(

        y_true == y_pred

    )



    incorrect_predictions = (

        total_predictions -

        correct_predictions

    )



    error_rate = (

        incorrect_predictions /

        total_predictions

    )



    
    # Create Summary Dictionary
  


    prediction_summary = {


        "Model":

        "MobileNetV2",



        "Total Predictions":

        int(total_predictions),



        "Correct Predictions":

        int(correct_predictions),



        "Incorrect Predictions":

        int(incorrect_predictions),



        "Accuracy":

        round(float(prediction_accuracy),4),



        "Precision":

        round(float(prediction_precision),4),



        "Recall":

        round(float(prediction_recall),4),



        "F1 Score":

        round(float(prediction_f1),4),



        "Error Rate":

        round(float(error_rate),4)

    }



    
    # Logging Results
  


    logger.info("------------------------------------------")


    logger.info(

        f"Total Predictions       : {total_predictions}"

    )


    logger.info(

        f"Correct Predictions     : {correct_predictions}"

    )


    logger.info(

        f"Incorrect Predictions   : {incorrect_predictions}"

    )


    logger.info("------------------------------------------")


    logger.info(

        f"Accuracy                : {prediction_accuracy:.4f}"

    )


    logger.info(

        f"Precision               : {prediction_precision:.4f}"

    )


    logger.info(

        f"Recall                  : {prediction_recall:.4f}"

    )


    logger.info(

        f"F1 Score                : {prediction_f1:.4f}"

    )


    logger.info("------------------------------------------")


    logger.info(

        f"Error Rate              : {error_rate:.4f}"

    )


    logger.info("------------------------------------------")


    logger.info(

        "Prediction results summary created successfully"

    )


    logger.info("="*70)



except Exception as e:


    logger.exception(

        f"Prediction summary generation failed: {str(e)}"

    )


    raise e
```

**7.Correct vs Incorrect Prediction Analysis**

**Implements:-**

```python



import logging
import pandas as pd
import numpy as np



logger = logging.getLogger(
    "RiceLeaf_MobileNetV2_Prediction"
)



try:


    logger.info("="*70)

    logger.info(
        "Running Correct vs Incorrect Prediction Analysis"
    )

    logger.info("="*70)



    
    # Validate Prediction Variables
    


    required_variables = [

        "y_true",

        "y_pred",

        "true_labels",

        "prediction_labels",

        "confidence_scores"

    ]


    for variable in required_variables:


        if variable not in globals():

            raise ValueError(

                f"{variable} not available"

            )



    
    # Identify Correct / Incorrect Predictions
    


    prediction_status = np.where(

        y_true == y_pred,

        "Correct",

        "Incorrect"

    )



   
    # Create Analysis DataFrame
    


    prediction_analysis_df = pd.DataFrame({


        "Actual_Label":

        true_labels,



        "Predicted_Label":

        prediction_labels,



        "Confidence":

        np.round(

            confidence_scores,

            4

        ),



        "Status":

        prediction_status

    })



   
    # Statistics
    


    total_predictions = len(

        prediction_status

    )


    correct_predictions = np.sum(

        prediction_status == "Correct"

    )


    incorrect_predictions = np.sum(

        prediction_status == "Incorrect"

    )



    accuracy_percentage = (

        correct_predictions /

        total_predictions

    ) * 100



    
    # Misclassified Samples
   


    misclassified_df = (

        prediction_analysis_df[

            prediction_analysis_df["Status"]

            ==

            "Incorrect"

        ]

    )



    
    # Class-wise Error Analysis
    


    error_analysis = pd.crosstab(

        prediction_analysis_df["Actual_Label"],

        prediction_analysis_df["Predicted_Label"]

    )



   
    # Logging Results
   


    logger.info("------------------------------------------")


    logger.info(

        f"Total Predictions      : {total_predictions}"

    )


    logger.info(

        f"Correct Predictions    : {correct_predictions}"

    )


    logger.info(

        f"Incorrect Predictions  : {incorrect_predictions}"

    )


    logger.info(

        f"Prediction Accuracy    : {accuracy_percentage:.2f}%"

    )


    logger.info("------------------------------------------")


    logger.info(

        f"Misclassified Samples  : {len(misclassified_df)}"

    )


    logger.info("------------------------------------------")


    logger.info(

        "Class-wise Error Matrix:"
    )


    print(error_analysis)



    logger.info("------------------------------------------")


    logger.info(

        "Correct vs Incorrect prediction analysis completed successfully"

    )


    logger.info("="*70)



except Exception as e:


    logger.exception(

        f"Prediction error analysis failed: {str(e)}"

    )


    raise e
```

**Implements:-**

```python
# ============================================================
# 15.2 GENERATE FINAL MOBILENETV2 TEST PREDICTIONS
# ============================================================


import logging
import numpy as np


logger = logging.getLogger(
    "RiceLeaf_MobileNetV2_Evaluation"
)


try:

    logger.info("=" * 70)

    logger.info(
        "Generating Final MobileNetV2 Predictions"
    )

    logger.info("=" * 70)



    # --------------------------------------------------------
    # Validate Required Objects
    # --------------------------------------------------------

    required_objects = [

        "mobilenet_model",
        "validation_generator"

    ]


    for obj in required_objects:

        if obj not in globals():

            raise ValueError(
                f"{obj} not found"
            )


    logger.info(
        "Required objects verified"
    )



    # --------------------------------------------------------
    # Reset Validation Generator
    # --------------------------------------------------------

    validation_generator.reset()


    logger.info(
        "Validation generator reset"
    )



    # --------------------------------------------------------
    # Generate Prediction Probabilities
    # --------------------------------------------------------

    prediction_probabilities = mobilenet_model.predict(

        validation_generator,

        verbose=1

    )



    # --------------------------------------------------------
    # Convert Probabilities to Class Labels
    # --------------------------------------------------------

    prediction_labels = np.argmax(

        prediction_probabilities,

        axis=1

    )


    true_labels = validation_generator.classes



    confidence_scores = np.max(

        prediction_probabilities,

        axis=1

    )



    class_names = list(

        validation_generator.class_indices.keys()

    )



    # --------------------------------------------------------
    # Prediction Validation
    # --------------------------------------------------------

    total_predictions = len(
        prediction_labels
    )


    total_ground_truth = len(
        true_labels
    )



    if total_predictions != total_ground_truth:


        raise ValueError(
            "Prediction count mismatch"
        )



    # --------------------------------------------------------
    # Logging Results
    # --------------------------------------------------------

    logger.info("------------------------------------------")


    logger.info(
        f"Prediction Samples     : {total_predictions}"
    )


    logger.info(
        f"Ground Truth Samples   : {total_ground_truth}"
    )


    logger.info(
        f"Prediction Shape       : {prediction_probabilities.shape}"
    )


    logger.info(
        f"Average Confidence     : {confidence_scores.mean():.4f}"
    )


    logger.info(
        f"Maximum Confidence     : {confidence_scores.max():.4f}"
    )


    logger.info(
        f"Minimum Confidence     : {confidence_scores.min():.4f}"
    )


    logger.info(
        f"Classes                : {class_names}"
    )


    logger.info("------------------------------------------")



    logger.info(
        "Final predictions generated successfully"
    )



except Exception as e:


    logger.exception(
        f"Prediction generation failed : {str(e)}"
    )


    raise
```

**3.Confusion Matrix Generation**

**Implements:-**

```python


import os
import logging
import numpy as np
import matplotlib.pyplot as plt

from sklearn.metrics import confusion_matrix


logger = logging.getLogger(
    "RiceLeaf_MobileNetV2_Evaluation"
)


try:

    logger.info("=" * 70)
    logger.info(
        "Generating MobileNetV2 Confusion Matrix"
    )
    logger.info("=" * 70)


   
    # Validate Required Variables
    

    required_variables = [

        "true_labels",

        "prediction_labels",

        "class_names",

        "evaluation_dir"

    ]


    for variable in required_variables:

        if variable not in globals():

            raise ValueError(
                f"{variable} not found"
            )


    logger.info(
        "Required evaluation variables verified"
    )


    
    # Generate Confusion Matrix
    

    cm = confusion_matrix(

        true_labels,

        prediction_labels

    )


   
    # Normalized Confusion Matrix
   

    cm_normalized = (

        cm.astype("float")

        /

        cm.sum(axis=1, keepdims=True)

    )


  
    # Plot
    

    fig, ax = plt.subplots(

        figsize=(8,6)

    )


    image = ax.imshow(

        cm,

        interpolation="nearest"

    )


    plt.colorbar(image)


    ax.set(

        xticks=np.arange(len(class_names)),

        yticks=np.arange(len(class_names)),

        xticklabels=class_names,

        yticklabels=class_names,

        xlabel="Predicted Label",

        ylabel="True Label",

        title="MobileNetV2 Confusion Matrix"

    )


    plt.setp(

        ax.get_xticklabels(),

        rotation=25,

        ha="right"

    )


    threshold = cm.max() / 2


    for i in range(cm.shape[0]):

        for j in range(cm.shape[1]):

            ax.text(

                j,

                i,

                f"{cm[i,j]}\n({cm_normalized[i,j]*100:.1f}%)",

                ha="center",

                va="center",

                color="white"

                if cm[i,j] > threshold

                else "black",

                fontsize=10

            )


    plt.tight_layout()


   
    # Save Figure
   

    confusion_matrix_path = os.path.join(

        evaluation_dir,

        "mobilenetv2_confusion_matrix.png"

    )


    plt.savefig(

        confusion_matrix_path,

        dpi=300,

        bbox_inches="tight"

    )


    plt.show()


   
    # Statistics
   

    correct_predictions = np.trace(cm)

    total_predictions = np.sum(cm)

    overall_accuracy = (

        correct_predictions

        /

        total_predictions

    )


    logger.info("------------------------------------------")

    logger.info(
        f"Correct Predictions : {correct_predictions}"
    )

    logger.info(
        f"Incorrect Predictions : {total_predictions-correct_predictions}"
    )

    logger.info(
        f"Overall Accuracy : {overall_accuracy:.4f}"
    )

    logger.info(
        f"Confusion Matrix Saved : {confusion_matrix_path}"
    )

    logger.info("------------------------------------------")


    logger.info(
        "Confusion matrix generated successfully"
    )

    logger.info("=" * 70)


except Exception as e:

    logger.exception(
        f"Confusion matrix generation failed: {str(e)}"
    )

    raise e
```

**6.Misclassification Analysis**

**Implements:-**

```python


import os
import logging
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt


logger = logging.getLogger(
    "RiceLeaf_MobileNetV2_Evaluation"
)


try:


    logger.info("=" * 70)

    logger.info(
        "Running MobileNetV2 Misclassification Analysis"
    )

    logger.info("=" * 70)



   
    # Validate Required Variables
   

    required_variables = [

        "true_labels",

        "prediction_labels",

        "confidence_scores",

        "class_names",

        "validation_generator",

        "evaluation_dir"

    ]


    for variable in required_variables:


        if variable not in globals():

            raise ValueError(

                f"{variable} not found"

            )


    logger.info(
        "Required variables verified"
    )



   
    # Identify Misclassified Samples
   

    misclassified_indices = np.where(

        true_labels != prediction_labels

    )[0]



    logger.info(
        "Misclassified samples identified"
    )



   
    # Create Misclassification DataFrame
    

    misclassification_records = []



    filepaths = validation_generator.filepaths



    for index in misclassified_indices:


        record = {


            "Image_Path":

            filepaths[index],



            "True_Label":

            class_names[

                true_labels[index]

            ],



            "Predicted_Label":

            class_names[

                prediction_labels[index]

            ],



            "Confidence":

            float(

                confidence_scores[index]

            )

        }


        misclassification_records.append(

            record

        )



    misclassification_df = pd.DataFrame(

        misclassification_records

    )



   
    # Save Misclassification Report
    

    misclassification_csv = os.path.join(

        evaluation_dir,

        "mobilenetv2_misclassification_analysis.csv"

    )


    misclassification_df.to_csv(

        misclassification_csv,

        index=False

    )



   
    # Visualization
   


    if len(misclassified_indices) > 0:


        plt.figure(

            figsize=(8,6)

        )


        for idx, sample_index in enumerate(

            misclassified_indices

        ):


            image = plt.imread(

                filepaths[sample_index]

            )


            plt.imshow(

                image

            )


            plt.axis("off")


            plt.title(

                f"Actual: {class_names[true_labels[sample_index]]}\n"

                f"Predicted: {class_names[prediction_labels[sample_index]]}\n"

                f"Confidence: {confidence_scores[sample_index]:.2%}"

            )



        misclassification_image_path = os.path.join(

            evaluation_dir,

            "mobilenetv2_misclassified_samples.png"

        )


        plt.savefig(

            misclassification_image_path,

            dpi=300,

            bbox_inches="tight"

        )


        plt.show()



    else:


        misclassification_image_path = "No misclassified images"



   
    # Logging Summary
   


    logger.info("------------------------------------------")


    logger.info(

        f"Total Samples          : {len(true_labels)}"

    )


    logger.info(

        f"Misclassified Samples  : {len(misclassified_indices)}"

    )


    logger.info(

        f"Analysis CSV Saved     : {misclassification_csv}"

    )


    logger.info(

        f"Visualization Saved    : {misclassification_image_path}"

    )


    logger.info("------------------------------------------")


    if len(misclassified_indices) > 0:


        logger.info(

            "\n%s",

            misclassification_df

        )


    logger.info(

        "Misclassification analysis completed successfully"

    )


    logger.info("=" * 70)



except Exception as e:


    logger.exception(

        f"Misclassification analysis failed: {str(e)}"

    )


    raise e
```

**2.Validate Model & Select Target Layer**

**Implements:-**

```python


base_model = None

for layer in mobilenet_model.layers:

    if isinstance(layer, tf.keras.Model):

        base_model = layer

        break

if base_model is None:

    raise ValueError(
        "MobileNetV2 backbone not found"
    )

logger.info(
    f"Backbone Model      : {base_model.name}"
)

logger.info(
    f"Backbone Layers     : {len(base_model.layers)}"
)


# Select Last Conv2D Layer


gradcam_target_layer = None

for layer in reversed(base_model.layers):

    if isinstance(layer, tf.keras.layers.Conv2D):

        gradcam_target_layer = layer

        break

if gradcam_target_layer is None:

    raise ValueError(
        "No Conv2D layer found inside MobileNetV2 backbone"
    )

gradcam_layer_name = gradcam_target_layer.name

logger.info(
    f"Selected Layer      : {gradcam_layer_name}"
)

logger.info(
    f"Layer Type          : {type(gradcam_target_layer).__name__}"
)

logger.info(
    f"Layer Output Shape  : {gradcam_target_layer.output.shape}"
)
```

**3. Load Sample Image**



```python


try:

    logger.info("=" * 70)
    logger.info(
        "Loading Sample Images for MobileNetV2 Grad-CAM"
    )
    logger.info("=" * 70)


    
    # Validate Required Variables
   

    if "validation_generator" not in globals():

        raise ValueError(
            "Validation generator not found"
        )


    logger.info(
        "Validation generator verified"
    )


    
    # Extract Validation Image Paths


    image_paths = validation_generator.filepaths


    if len(image_paths) == 0:

        raise ValueError(
            "No validation images available"
        )


    logger.info(
        f"Total Validation Images : {len(image_paths)}"
    )


   
    # Select Misclassified Image
    

    misclassified_image_path = (
        r"dataset\Bacterial leaf blight\DSC_0374.JPG"
    )


    if misclassified_image_path not in image_paths:

        logger.warning(
            "Misclassified image not found in validation set"
        )


   
    # Select Correct Prediction Image Automatically
   

    correct_image_path = None


    if "y_true" in globals() and "y_pred_classes" in globals():

        for idx, (true_label, pred_label) in enumerate(
            zip(y_true, y_pred_classes)
        ):

            if true_label == pred_label:

                correct_image_path = image_paths[idx]

                break


    if correct_image_path is None:

        correct_image_path = image_paths[0]


   
    # Load Images
    

    from tensorflow.keras.utils import load_img


    correct_image = load_img(
        correct_image_path
    )


    misclassified_image = load_img(
        misclassified_image_path
    )


  
    # Store Metadata
   

    gradcam_correct_image_path = correct_image_path

    gradcam_wrong_image_path = misclassified_image_path


    logger.info("------------------------------------------")

    logger.info(
        f"Correct Image      : {gradcam_correct_image_path}"
    )

    logger.info(
        f"Misclassified Image: {gradcam_wrong_image_path}"
    )

    logger.info("------------------------------------------")


    logger.info(
        "Sample images loaded successfully"
    )

    logger.info("=" * 70)


except Exception as e:


    logger.exception(
        f"Sample image loading failed: {str(e)}"
    )

    raise e
```

**4.Image Preprocessing**


```python

try:

    logger.info("=" * 70)
    logger.info(
        "Preprocessing Images for MobileNetV2 Grad-CAM"
    )
    logger.info("=" * 70)


   
    # Validate Image Paths
   

    if "gradcam_correct_image_path" not in globals():

        raise ValueError(
            "Correct image path not found"
        )


    if "gradcam_wrong_image_path" not in globals():

        raise ValueError(
            "Misclassified image path not found"
        )


    logger.info(
        "Image paths verified"
    )


   
    

    from tensorflow.keras.utils import load_img, img_to_array
    from tensorflow.keras.applications.mobilenet_v2 import preprocess_input


   
    # Image Preprocessing Function
   

    def preprocess_gradcam_image(image_path):

        image = load_img(

            image_path,

            target_size=IMAGE_SIZE

        )


        image_array = img_to_array(

            image

        )


        original_image = image_array.copy()


        image_array = np.expand_dims(

            image_array,

            axis=0

        )


        processed_image = preprocess_input(

            image_array

        )


        return (
            original_image,
            processed_image
        )


    
    # Process Correct Image

    correct_original_image, correct_processed_image = (
        preprocess_gradcam_image(
            gradcam_correct_image_path
        )
    )


   
    # Process Misclassified Image#

    wrong_original_image, wrong_processed_image = (
        preprocess_gradcam_image(
            gradcam_wrong_image_path
        )
    )


   
    # Validate Shapes
    

    logger.info("------------------------------------------")

    logger.info(
        f"Correct Image Tensor Shape : {correct_processed_image.shape}"
    )

    logger.info(
        f"Wrong Image Tensor Shape   : {wrong_processed_image.shape}"
    )


    logger.info(
        f"Original Image Shape       : {correct_original_image.shape}"
    )


    logger.info("------------------------------------------")


    
    # Final Variables
   

    gradcam_correct_input = correct_processed_image

    gradcam_wrong_input = wrong_processed_image


    logger.info(
        "Image preprocessing completed successfully"
    )

    logger.info("=" * 70)


except Exception as e:

    logger.exception(
        f"Grad-CAM image preprocessing failed: {str(e)}"
    )

    raise e
```

**5.Generate Grad-CAM Heatmap**

**Implements:-**

```python


try:

    logger.info("=" * 70)
    logger.info(
        "Generating MobileNetV2 Grad-CAM Heatmaps"
    )
    logger.info("=" * 70)


    
    # Validate Variables
   

    if "mobilenet_model" not in globals():

        raise ValueError(
            "MobileNetV2 model not found"
        )


    if "gradcam_correct_input" not in globals():

        raise ValueError(
            "Correct input image not found"
        )


    if "gradcam_wrong_input" not in globals():

        raise ValueError(
            "Wrong input image not found"
        )


    logger.info(
        "Required Grad-CAM variables verified"
    )


    
    # Extract Backbone
   

    backbone_model = mobilenet_model.layers[1]


    logger.info(
        f"Backbone Model : {backbone_model.name}"
    )


   
    # Create Feature Extractor
  

    feature_extractor = tf.keras.Model(

        inputs=backbone_model.input,

        outputs=[

            gradcam_target_layer.output,

            backbone_model.output

        ]

    )


    logger.info(
        "Feature extractor created successfully"
    )


   
    # Grad-CAM Function
   
    def generate_gradcam(image_tensor):


        with tf.GradientTape() as tape:


            conv_output, backbone_output = feature_extractor(

                image_tensor,

                training=False

            )


            tape.watch(conv_output)


            x = backbone_output


            # Continue outer classifier

            for layer in mobilenet_model.layers[2:]:

                x = layer(x)


            predictions = x


            predicted_class = tf.argmax(

                predictions[0]

            )


            loss = predictions[:, predicted_class]



        gradients = tape.gradient(

            loss,

            conv_output

        )


        if gradients is None:

            raise ValueError(
                "Gradient calculation failed"
            )


        pooled_gradients = tf.reduce_mean(

            gradients,

            axis=(0,1,2)

        )


        conv_output = conv_output[0]


        heatmap = tf.reduce_sum(

            conv_output *
            pooled_gradients,

            axis=-1

        )


        heatmap = tf.maximum(

            heatmap,

            0

        )


        heatmap /= (

            tf.reduce_max(heatmap)

            + 1e-10

        )


        return (

            heatmap.numpy(),

            int(predicted_class.numpy()),

            predictions.numpy()

        )


    
    # Correct Image
    
    (
        gradcam_correct_heatmap,
        gradcam_correct_class,
        gradcam_correct_prediction

    ) = generate_gradcam(

        gradcam_correct_input

    )


   
    # Wrong Image
   

    (
        gradcam_wrong_heatmap,
        gradcam_wrong_class,
        gradcam_wrong_prediction

    ) = generate_gradcam(

        gradcam_wrong_input

    )


    logger.info("------------------------------------------")

    logger.info(
        f"Correct Prediction Class : {gradcam_correct_class}"
    )


    logger.info(
        f"Wrong Prediction Class   : {gradcam_wrong_class}"
    )


    logger.info(
        f"Correct Heatmap Shape    : {gradcam_correct_heatmap.shape}"
    )


    logger.info(
        f"Wrong Heatmap Shape      : {gradcam_wrong_heatmap.shape}"
    )


    logger.info("------------------------------------------")


    logger.info(
        "Grad-CAM heatmaps generated successfully"
    )


    logger.info("=" * 70)



except Exception as e:

    logger.exception(
        f"Grad-CAM generation failed: {str(e)}"
    )

    raise e
```

**6.Heatmap Post-processing & Normalization**


```python


try:

    logger.info("=" * 70)
    logger.info(
        "Processing MobileNetV2 Grad-CAM Heatmaps"
    )
    logger.info("=" * 70)


    
    # Validate Heatmaps
    
    if "gradcam_correct_heatmap" not in globals():

        raise ValueError(
            "Correct Grad-CAM heatmap not found"
        )


    if "gradcam_wrong_heatmap" not in globals():

        raise ValueError(
            "Wrong Grad-CAM heatmap not found"
        )


    logger.info(
        "Grad-CAM heatmaps verified"
    )


    
    # Import Required Libraries
 

    import cv2
    import numpy as np


   
    # Heatmap Processing Function
   

    def process_heatmap(
        heatmap,
        target_size
    ):


        # Resize heatmap

        heatmap = cv2.resize(

            heatmap,

            target_size,

            interpolation=cv2.INTER_LINEAR

        )


        # Normalize 0-1

        heatmap = np.maximum(

            heatmap,

            0

        )


        heatmap = heatmap / (

            np.max(heatmap)

            + 1e-10

        )


        # Convert to 0-255

        heatmap_uint8 = np.uint8(

            255 * heatmap

        )


        return heatmap_uint8



    
    # Process Correct Heatmap
  

    processed_correct_heatmap = process_heatmap(

        gradcam_correct_heatmap,

        (
            IMAGE_SIZE[1],

            IMAGE_SIZE[0]

        )

    )


   
    # Process Wrong Heatmap
  

    processed_wrong_heatmap = process_heatmap(

        gradcam_wrong_heatmap,

        (
            IMAGE_SIZE[1],

            IMAGE_SIZE[0]

        )

    )


    
    # Store Variables
    

    gradcam_correct_processed_heatmap = (
        processed_correct_heatmap
    )


    gradcam_wrong_processed_heatmap = (
        processed_wrong_heatmap
    )


    
    # Validation Logs
    

    logger.info("------------------------------------------")


    logger.info(
        f"Raw Correct Heatmap Shape  : {gradcam_correct_heatmap.shape}"
    )


    logger.info(
        f"Raw Wrong Heatmap Shape    : {gradcam_wrong_heatmap.shape}"
    )


    logger.info(
        f"Processed Correct Shape    : {processed_correct_heatmap.shape}"
    )


    logger.info(
        f"Processed Wrong Shape      : {processed_wrong_heatmap.shape}"
    )


    logger.info(
        f"Correct Heatmap Min Value  : {processed_correct_heatmap.min()}"
    )


    logger.info(
        f"Correct Heatmap Max Value  : {processed_correct_heatmap.max()}"
    )


    logger.info(
        f"Wrong Heatmap Min Value    : {processed_wrong_heatmap.min()}"
    )


    logger.info(
        f"Wrong Heatmap Max Value    : {processed_wrong_heatmap.max()}"
    )


    logger.info("------------------------------------------")


    logger.info(
        "Heatmap post-processing completed successfully"
    )


    logger.info("=" * 70)



except Exception as e:


    logger.exception(
        f"Heatmap processing failed: {str(e)}"
    )

    raise e
```

**7.Overlay Heatmap**

```python

# Grad-CAM Directory Configuration Fix


import os

GRADCAM_DIR = os.path.join(
    "artifacts",
    "gradcam"
)

os.makedirs(
    GRADCAM_DIR,
    exist_ok=True
)


logger.info(
    f"Grad-CAM Directory Ready : {GRADCAM_DIR}"
)
```

```python


try:

    logger.info("=" * 70)
    logger.info(
        "Generating MobileNetV2 Grad-CAM Overlay Images"
    )
    logger.info("=" * 70)


    
    # Validate Required Variables
   

    required_variables = [

        "gradcam_correct_processed_heatmap",
        "gradcam_wrong_processed_heatmap",
        "correct_original_image",
        "wrong_original_image"

    ]


    for variable in required_variables:

        if variable not in globals():

            raise ValueError(
                f"{variable} not found"
            )


    logger.info(
        "Required Grad-CAM overlay variables verified"
    )


    import cv2
    import numpy as np


    
    # Overlay Function
  

    def create_gradcam_overlay(

        original_image,

        heatmap,

        alpha=0.4

    ):


        # Convert RGB to uint8

        original_image = np.uint8(

            original_image * 255

        )


        # Generate color heatmap

        heatmap_color = cv2.applyColorMap(

            heatmap,

            cv2.COLORMAP_JET

        )


        # Convert BGR to RGB

        heatmap_color = cv2.cvtColor(

            heatmap_color,

            cv2.COLOR_BGR2RGB

        )


        # Weighted combination

        overlay = cv2.addWeighted(

            original_image,

            1-alpha,

            heatmap_color,

            alpha,

            0

        )


        return overlay



   
    # Generate Correct Overlay
   
    correct_overlay = create_gradcam_overlay(

        correct_original_image,

        gradcam_correct_processed_heatmap

    )


   
    # Generate Wrong Overlay
   

    wrong_overlay = create_gradcam_overlay(

        wrong_original_image,

        gradcam_wrong_processed_heatmap

    )


    
    # Store Results
   

    gradcam_correct_overlay = correct_overlay

    gradcam_wrong_overlay = wrong_overlay



    
    # Save Directory
  

    import os


    os.makedirs(

        GRADCAM_DIR,

        exist_ok=True

    )


    correct_path = os.path.join(

        GRADCAM_DIR,

        "correct_prediction_gradcam_overlay.png"

    )


    wrong_path = os.path.join(

        GRADCAM_DIR,

        "misclassified_prediction_gradcam_overlay.png"

    )


    cv2.imwrite(

        correct_path,

        cv2.cvtColor(

            correct_overlay,

            cv2.COLOR_RGB2BGR

        )

    )


    cv2.imwrite(

        wrong_path,

        cv2.cvtColor(

            wrong_overlay,

            cv2.COLOR_RGB2BGR

        )

    )


   
    # Logs
   

    logger.info("------------------------------------------")


    logger.info(
        f"Correct Overlay Shape : {correct_overlay.shape}"
    )


    logger.info(
        f"Wrong Overlay Shape   : {wrong_overlay.shape}"
    )


    logger.info(
        f"Correct Overlay Saved : {correct_path}"
    )


    logger.info(
        f"Wrong Overlay Saved   : {wrong_path}"
    )


    logger.info("------------------------------------------")


    logger.info(
        "Grad-CAM overlay generation completed successfully"
    )


    logger.info("=" * 70)



except Exception as e:


    logger.exception(

        f"Grad-CAM overlay generation failed: {str(e)}"

    )

    raise e
```

**7.Model Loading Verification**

```python

import os
import tensorflow as tf


try:

    logger.info("=" * 70)
    logger.info(
        "Loading Saved MobileNetV2 Model Verification"
    )
    logger.info("=" * 70)



   
    # Model Path
    

    MODEL_PATH = (
        "artifacts/model_registry/"
        "RiceLeaf_MobileNetV2_v1.0.keras"
    )


    if not os.path.exists(MODEL_PATH):

        raise FileNotFoundError(
            "Saved model file not found"
        )


    logger.info(
        "Saved model file verified"
    )



    
    # Load Model
    

    loaded_mobilenet_model = tf.keras.models.load_model(
        MODEL_PATH
    )


    logger.info(
        "Model loaded successfully"
    )



   
    # Architecture Verification
   

    original_input_shape = (
        mobilenet_model.input_shape
    )

    loaded_input_shape = (
        loaded_mobilenet_model.input_shape
    )


    original_output_shape = (
        mobilenet_model.output_shape
    )

    loaded_output_shape = (
        loaded_mobilenet_model.output_shape
    )



    input_match = (
        original_input_shape ==
        loaded_input_shape
    )


    output_match = (
        original_output_shape ==
        loaded_output_shape
    )



    parameter_match = (

        mobilenet_model.count_params()
        ==
        loaded_mobilenet_model.count_params()

    )



    logger.info("-" * 42)


    logger.info(
        f"Input Shape Match    : {input_match}"
    )


    logger.info(
        f"Output Shape Match   : {output_match}"
    )


    logger.info(
        f"Parameter Match      : {parameter_match}"
    )


    logger.info(
        f"Original Parameters  : {mobilenet_model.count_params()}"
    )


    logger.info(
        f"Loaded Parameters    : {loaded_mobilenet_model.count_params()}"
    )


    logger.info("-" * 42)



  
    # Final Validation
   

    if (

        input_match
        and output_match
        and parameter_match

    ):

        logger.info(
            "Model Loading Verification : SUCCESS"
        )


        logger.info(
            "Loaded Model Ready For Inference"
        )


    else:

        raise ValueError(
            "Loaded model validation failed"
        )


    logger.info("=" * 70)



except Exception as e:


    logger.exception(
        f"Model loading verification failed : {str(e)}"
    )


    raise e
```

**8.Prediction Verification After Reload**

```python


import numpy as np
import tensorflow as tf


try:

    logger.info("=" * 70)
    logger.info(
        "Verifying Prediction After MobileNetV2 Model Reload"
    )
    logger.info("=" * 70)



    
    # Validate Required Objects
   

    required_objects = [

        "mobilenet_model",
        "loaded_mobilenet_model",
        "gradcam_correct_input",
        "class_names"

    ]


    for obj in required_objects:

        if obj not in globals():

            raise ValueError(
                f"{obj} not found"
            )


    logger.info(
        "Required objects verified"
    )



    
    # Original Model Prediction
 

    original_prediction = (
        mobilenet_model.predict(
            gradcam_correct_input,
            verbose=0
        )
    )


    original_class_index = int(
        np.argmax(original_prediction)
    )


    original_confidence = float(
        np.max(original_prediction)
    )



   
    # Reloaded Model Prediction
    

    loaded_prediction = (
        loaded_mobilenet_model.predict(
            gradcam_correct_input,
            verbose=0
        )
    )


    loaded_class_index = int(
        np.argmax(loaded_prediction)
    )


    loaded_confidence = float(
        np.max(loaded_prediction)
    )



    
    # Compare Predictions
    

    class_match = (

        original_class_index
        ==
        loaded_class_index

    )


    confidence_difference = abs(

        original_confidence
        -
        loaded_confidence

    )



    logger.info("-" * 42)


    logger.info(
        f"Original Class Index  : {original_class_index}"
    )


    logger.info(
        f"Reloaded Class Index  : {loaded_class_index}"
    )


    logger.info(
        f"Original Class Name   : {class_names[original_class_index]}"
    )


    logger.info(
        f"Reloaded Class Name   : {class_names[loaded_class_index]}"
    )


    logger.info(
        f"Original Confidence   : {original_confidence:.4f}"
    )


    logger.info(
        f"Reloaded Confidence   : {loaded_confidence:.4f}"
    )


    logger.info(
        f"Confidence Difference : {confidence_difference:.8f}"
    )


    logger.info(
        f"Prediction Match      : {class_match}"
    )


    logger.info("-" * 42)



   
    # Final Validation
    

    if class_match and confidence_difference < 1e-5:


        logger.info(
            "Prediction Verification After Reload : SUCCESS"
        )


        logger.info(
            "Saved Model Produces Identical Output"
        )


    else:


        raise ValueError(
            "Prediction mismatch after reload"
        )



    logger.info("=" * 70)



except Exception as e:


    logger.exception(
        f"Prediction verification failed : {str(e)}"
    )


    raise e
```

**9.Final Model Package Validation**

```python


import os
import json


try:

    logger.info("=" * 70)
    logger.info(
        "Running Final MobileNetV2 Model Package Validation"
    )
    logger.info("=" * 70)



   
    # Model Registry Directory
    

    MODEL_DIR = "artifacts/model_registry"



    
    # Required Files
    

    required_files = {

        "Keras Model":
        "RiceLeaf_MobileNetV2_v1.0.keras",

        "Metadata JSON":
        "RiceLeaf_MobileNetV2_v1.0_metadata.json",

        "Class Labels":
        "RiceLeaf_MobileNetV2_v1.0_class_labels.json",

        "Training Config":
        "RiceLeaf_MobileNetV2_v1.0_training_config.json",

        "Version Config":
        "version_config.json"

    }



    validation_results = {}



   
    # File Validation
   

    for name, file in required_files.items():

        path = os.path.join(
            MODEL_DIR,
            file
        )

        validation_results[name] = os.path.exists(path)



    
    # Object Validation
  

    validation_results["Final Model Available"] = (

        "mobilenet_model" in globals()

    )


    validation_results["Reloaded Model Available"] = (

        "loaded_mobilenet_model" in globals()

    )



    validation_results["Prediction Verification"] = (

        "class_match" in globals()
        and class_match is True

    )



    
    # JSON Validation
   

    json_files = [

        "RiceLeaf_MobileNetV2_v1.0_metadata.json",

        "RiceLeaf_MobileNetV2_v1.0_class_labels.json",

        "RiceLeaf_MobileNetV2_v1.0_training_config.json",

        "version_config.json"

    ]



    for file in json_files:

        path = os.path.join(
            MODEL_DIR,
            file
        )


        try:

            with open(
                path,
                "r"
            ) as f:

                json.load(f)


            validation_results[
                f"{file} Readable"
            ] = True


        except Exception:

            validation_results[
                f"{file} Readable"
            ] = False



   
    # Logging Results
    

    logger.info("-" * 42)


    failed_checks = []


    for check, status in validation_results.items():

        logger.info(
            f"{check:<35}: {status}"
        )


        if status is False:

            failed_checks.append(check)



    logger.info("-" * 42)



   
    # Final Status
   
    if len(failed_checks) == 0:


        logger.info(
            "Total Validation Checks Passed : "
            f"{len(validation_results)}"
        )


        logger.info(
            "Final Model Package Status : SUCCESS"
        )


        logger.info(
            "MobileNetV2 Model Ready For Deployment"
        )


    else:


        logger.error(
            f"Failed Checks : {failed_checks}"
        )


        raise ValueError(
            "Final Model Package Validation Failed"
        )



    logger.info("=" * 70)



except Exception as e:


    logger.exception(
        f"Model Package Validation Failed : {str(e)}"
    )


    raise e
```

```python

try:

    logger.info("=" * 70)
    logger.info("Validating Input Image File")
    logger.info("=" * 70)

   
    # Required Libraries
    
    import os
    from pathlib import Path

    
    # Input Image Path
   

    INPUT_IMAGE_PATH = correct_image_path

    logger.info(
        "Input image path received"
    )

    
    # Validation Checks
 

    file_exists = os.path.exists(
        INPUT_IMAGE_PATH
    )

    if not file_exists:

        raise FileNotFoundError(
            f"File not found : {INPUT_IMAGE_PATH}"
        )

    is_file = os.path.isfile(
        INPUT_IMAGE_PATH
    )

    if not is_file:

        raise ValueError(
            "Input path is not a valid file"
        )

    is_readable = os.access(
        INPUT_IMAGE_PATH,
        os.R_OK
    )

    if not is_readable:

        raise PermissionError(
            "Input file is not readable"
        )

    file_size_bytes = os.path.getsize(
        INPUT_IMAGE_PATH
    )

    file_size_mb = round(

        file_size_bytes /
        (1024 * 1024),

        4

    )

    file_not_empty = (

        file_size_bytes > 0

    )

    file_size_valid = (

        file_size_bytes <= MAX_FILE_SIZE_BYTES

    )

    file_extension = Path(
        INPUT_IMAGE_PATH
    ).suffix.lower()

    extension_available = (

        file_extension != ""

    )

   
    # Validation Status
   
    validation_status = (

        file_exists and
        is_file and
        is_readable and
        file_not_empty and
        file_size_valid and
        extension_available

    )

    
   

    input_file_metadata = {

        "file_name":

            Path(
                INPUT_IMAGE_PATH
            ).name,

        "file_path":

            str(
                INPUT_IMAGE_PATH
            ),

        "extension":

            file_extension,

        "file_size_bytes":

            int(
                file_size_bytes
            ),

        "file_size_mb":

            float(
                file_size_mb
            )

    }

    
    # Logging Summary
    

    logger.info("-" * 42)

    logger.info(

        f"Input File            : "
        f"{input_file_metadata['file_name']}"

    )

    logger.info(

        f"Extension             : "
        f"{file_extension}"

    )

    logger.info(

        f"File Size             : "
        f"{file_size_mb:.4f} MB"

    )

    logger.info("-" * 42)

    logger.info(

        f"File Exists           : {file_exists}"

    )

    logger.info(

        f"Valid File            : {is_file}"

    )

    logger.info(

        f"Readable              : {is_readable}"

    )

    logger.info(

        f"File Not Empty        : {file_not_empty}"

    )

    logger.info(

        f"File Size Valid       : {file_size_valid}"

    )

    logger.info(

        f"Extension Available   : {extension_available}"

    )

    logger.info("-" * 42)

    logger.info(

        f"Overall Validation    : {validation_status}"

    )

    if not validation_status:

        raise ValueError(
            "Input File Validation Failed"
        )

    logger.info(
        "Input File Validation Completed Successfully"
    )

    logger.info("=" * 70)

except Exception as e:

    logger.exception(
        f"Input file validation failed : {str(e)}"
    )

    raise
```

**3.Validate Image Format**

```python


try:

    logger.info("=" * 70)
    logger.info("Validating Input Image Format")
    logger.info("=" * 70)


  
    # Required Libraries
   

    import os
    from pathlib import Path
    from PIL import Image


    
    

    SUPPORTED_FORMATS = [

        ".jpg",
        ".jpeg",
        ".png",
        ".bmp",
        ".tif",
        ".tiff"

    ]


    
    # Validate Required Input
   

    if "INPUT_IMAGE_PATH" not in globals():

        raise ValueError(
            "INPUT_IMAGE_PATH not found"
        )


    logger.info(
        "Required variables verified"
    )


    # -------------------------------------------------------------------------
    # Validate Image Opening
    # -------------------------------------------------------------------------

    try:

        image = Image.open(
            INPUT_IMAGE_PATH
        )


        detected_pil_format = image.format


        image.verify()


        logger.info(
            "Image decoding successful"
        )


    except Exception as e:

        raise ValueError(
            f"Corrupted image detected : {str(e)}"
        )


    
    # Reopen Image After Verification


    image = Image.open(
        INPUT_IMAGE_PATH
    )


    
    # Extract Metadata
    
    detected_format = (

        "." +
        detected_pil_format.lower()

    )


    file_extension = (

        Path(
            INPUT_IMAGE_PATH
        )
        .suffix
        .lower()

    )


    image_mode = image.mode


    format_supported = (

        detected_format in SUPPORTED_FORMATS

    )


    extension_match = (

        detected_format ==
        file_extension

    )


    image_readable = True


   
    # Save Format Metadata
   

    image_format_metadata = {


        "file_name":

            Path(
                INPUT_IMAGE_PATH
            ).name,


        "detected_format":

            detected_format,


        "file_extension":

            file_extension,


        "image_mode":

            image_mode,


        "format_supported":

            format_supported,


        "extension_match":

            extension_match,


        "readable":

            image_readable

    }


    
    # Final Validation
   

    format_validation_status = (

        format_supported
        and
        image_readable

    )


    
    # Logging
   

    logger.info("-" * 42)


    logger.info(
        f"Detected Format        : {detected_format}"
    )


    logger.info(
        f"File Extension         : {file_extension}"
    )


    logger.info(
        f"Image Mode             : {image_mode}"
    )


    logger.info(
        f"Supported Format       : {format_supported}"
    )


    logger.info(
        f"Extension Match        : {extension_match}"
    )


    logger.info(
        f"Image Readable         : {image_readable}"
    )


    logger.info("-" * 42)


    logger.info(
        f"Format Validation      : {format_validation_status}"
    )


    if not format_validation_status:

        raise ValueError(
            "Image Format Validation Failed"
        )


    logger.info(
        "Image Format Validation Completed Successfully"
    )


    logger.info("=" * 70)



except Exception as e:


    logger.exception(
        f"Image format validation failed : {str(e)}"
    )

    raise
```

**4.Validate Image Dimensions**

```python


try:

    logger.info("=" * 70)
    logger.info("Validating Image Dimensions")
    logger.info("=" * 70)


   
    # Required Libraries
   

    from PIL import Image


   
    # Configuration
  

    MIN_WIDTH = 64
    MIN_HEIGHT = 64

    MAX_WIDTH = 4096
    MAX_HEIGHT = 4096

    TARGET_SIZE = (
        224,
        224
    )


    
    # Validate Required Variables
    

    if "INPUT_IMAGE_PATH" not in globals():

        raise ValueError(
            "INPUT_IMAGE_PATH not found"
        )


    logger.info(
        "Required variables verified"
    )


   
    # Load Image
   

    image = Image.open(
        INPUT_IMAGE_PATH
    )


    
    # Extract Dimensions
  

    width, height = image.size


    aspect_ratio = round(

        width / height,

        4

    )


   
    # Dimension Validation
  

    minimum_dimension_valid = (

        width >= MIN_WIDTH
        and
        height >= MIN_HEIGHT

    )


    maximum_dimension_valid = (

        width <= MAX_WIDTH
        and
        height <= MAX_HEIGHT

    )


    dimension_valid = (

        minimum_dimension_valid
        and
        maximum_dimension_valid

    )


   
    # Resize Requirement
    

    resize_required = (

        width != TARGET_SIZE[0]
        or
        height != TARGET_SIZE[1]

    )


    
    # Store Metadata
   
    image_dimension_metadata = {


        "original_width":

            int(width),


        "original_height":

            int(height),


        "aspect_ratio":

            float(aspect_ratio),


        "minimum_dimension_valid":

            bool(minimum_dimension_valid),


        "maximum_dimension_valid":

            bool(maximum_dimension_valid),


        "resize_required":

            bool(resize_required),


        "target_size":

            list(TARGET_SIZE)

    }


   
    # Logging
   
    logger.info("-" * 42)


    logger.info(
        f"Image Width              : {width}px"
    )


    logger.info(
        f"Image Height             : {height}px"
    )


    logger.info(
        f"Aspect Ratio             : {aspect_ratio}"
    )


    logger.info(
        f"Minimum Dimension Check  : {minimum_dimension_valid}"
    )


    logger.info(
        f"Maximum Dimension Check  : {maximum_dimension_valid}"
    )


    logger.info(
        f"Resize Required          : {resize_required}"
    )


    logger.info("-" * 42)


    logger.info(
        f"Dimension Validation     : {dimension_valid}"
    )


    if not dimension_valid:

        raise ValueError(
            "Image Dimension Validation Failed"
        )


    logger.info(
        "Image Dimension Validation Completed Successfully"
    )


    logger.info("=" * 70)



except Exception as e:


    logger.exception(
        f"Image dimension validation failed : {str(e)}"
    )

    raise
```

**5.Validate Image Channels**

```python


try:

    logger.info("=" * 70)
    logger.info("Validating Image Channels")
    logger.info("=" * 70)


    
    # Required Libraries
    

    from PIL import Image


    
    # Configuration
   

    REQUIRED_CHANNELS = 3


   
    # Validate Required Variables
   

    if "INPUT_IMAGE_PATH" not in globals():

        raise ValueError(
            "INPUT_IMAGE_PATH not found"
        )


    logger.info(
        "Required variables verified"
    )


 
    # Load Image
    

    image = Image.open(
        INPUT_IMAGE_PATH
    )


    
    # Extract Image Mode
  

    image_mode = image.mode


   
    # Channel Detection
   

    channel_mapping = {


        "1": 1,        # Grayscale


        "L": 1,        # Grayscale


        "LA": 2,       # Grayscale + Alpha


        "RGB": 3,      # Standard Image


        "RGBA": 4      # RGB + Alpha

    }


    detected_channels = channel_mapping.get(

        image_mode,

        len(image.getbands())

    )


   
    # Channel Validation
  

    channel_count_valid = (

        detected_channels == REQUIRED_CHANNELS

    )


    
    # Conversion Requirement
    

    conversion_required = (

        detected_channels != REQUIRED_CHANNELS

    )


    
    # Alpha Channel Detection
    
    alpha_channel_present = (

        "A" in image_mode

    )


    
    # Store Metadata
    

    image_channel_metadata = {


        "image_mode":

            image_mode,


        "detected_channels":

            int(detected_channels),


        "required_channels":

            REQUIRED_CHANNELS,


        "channel_validation":

            bool(channel_count_valid),


        "alpha_channel_present":

            bool(alpha_channel_present),


        "conversion_required":

            bool(conversion_required)

    }


   
    # Logging
   

    logger.info("-" * 42)


    logger.info(
        f"Image Mode              : {image_mode}"
    )


    logger.info(
        f"Detected Channels       : {detected_channels}"
    )


    logger.info(
        f"Required Channels       : {REQUIRED_CHANNELS}"
    )


    logger.info(
        f"Alpha Channel Present   : {alpha_channel_present}"
    )


    logger.info(
        f"Conversion Required     : {conversion_required}"
    )


    logger.info("-" * 42)


    logger.info(
        f"Channel Validation      : {channel_count_valid}"
    )


    if not channel_count_valid:

        logger.warning(
            "Image channels require conversion before inference"
        )


    logger.info(
        "Image Channel Validation Completed Successfully"
    )


    logger.info("=" * 70)



except Exception as e:


    logger.exception(
        f"Image channel validation failed : {str(e)}"
    )

    raise
```

**6.Validate Image Integrity**

```python


try:

    logger.info("=" * 70)
    logger.info("Validating Image Integrity")
    logger.info("=" * 70)


    
    # Required Libraries
  

    import numpy as np
    from PIL import Image


   
    # Validate Required Variables
   

    if "INPUT_IMAGE_PATH" not in globals():

        raise ValueError(
            "INPUT_IMAGE_PATH not found"
        )


    logger.info(
        "Required variables verified"
    )


   
    # Load Image Safely
   

    try:

        image = Image.open(
            INPUT_IMAGE_PATH
        )


        image.load()


        logger.info(
            "Image pixel loading successful"
        )


    except Exception as e:

        raise ValueError(
            f"Image decoding failed : {str(e)}"
        )


   
    # Convert Image To Array
    

    image_array = np.array(
        image
    )


    
   

    image_not_empty = (

        image_array.size > 0

    )


    valid_shape = (

        len(image_array.shape) >= 2

    )


    has_nan_values = np.isnan(

        image_array.astype(
            "float32"
        )

    ).any()


    has_inf_values = np.isinf(

        image_array.astype(
            "float32"
        )

    ).any()


    pixel_range_valid = (

        image_array.min() >= 0

        and

        image_array.max() <= 255

    )


    
    # Final Integrity Status
   

    integrity_status = (

        image_not_empty

        and

        valid_shape

        and

        not has_nan_values

        and

        not has_inf_values

        and

        pixel_range_valid

    )


   
    # Save Metadata
   
    image_integrity_metadata = {


        "image_shape":

            list(image_array.shape),


        "image_dtype":

            str(image_array.dtype),


        "total_pixels":

            int(image_array.size),


        "minimum_pixel_value":

            int(image_array.min()),


        "maximum_pixel_value":

            int(image_array.max()),


        "contains_nan":

            bool(has_nan_values),


        "contains_infinity":

            bool(has_inf_values),


        "pixel_range_valid":

            bool(pixel_range_valid),


        "integrity_status":

            bool(integrity_status)

    }


    
    # Logging
  

    logger.info("-" * 42)


    logger.info(
        f"Image Shape              : {image_array.shape}"
    )


    logger.info(
        f"Image Data Type          : {image_array.dtype}"
    )


    logger.info(
        f"Total Pixels             : {image_array.size}"
    )


    logger.info(
        f"Minimum Pixel Value      : {image_array.min()}"
    )


    logger.info(
        f"Maximum Pixel Value      : {image_array.max()}"
    )


    logger.info(
        f"Contains NaN             : {has_nan_values}"
    )


    logger.info(
        f"Contains Infinity        : {has_inf_values}"
    )


    logger.info(
        f"Pixel Range Valid        : {pixel_range_valid}"
    )


    logger.info("-" * 42)


    logger.info(
        f"Image Integrity Status   : {integrity_status}"
    )


    if not integrity_status:

        raise ValueError(
            "Image Integrity Validation Failed"
        )


    logger.info(
        "Image Integrity Validation Completed Successfully"
    )


    logger.info("=" * 70)



except Exception as e:


    logger.exception(
        f"Image integrity validation failed : {str(e)}"
    )

    raise
```

**7.Preprocess Validated Image**

```python


try:

    logger.info("=" * 70)
    logger.info("Preprocessing Validated Image For MobileNetV2")
    logger.info("=" * 70)


   
    # Required Libraries
    

    import numpy as np
    from PIL import Image


    
    # Configuration
   

    MODEL_INPUT_SIZE = (
        224,
        224
    )


    NORMALIZATION_RANGE = (
        0.0,
        1.0
    )


   
    # Validate Input
    
    if "INPUT_IMAGE_PATH" not in globals():

        raise ValueError(
            "INPUT_IMAGE_PATH not found"
        )


    logger.info(
        "Required variables verified"
    )


    
    # Load Image
   
    image = Image.open(
        INPUT_IMAGE_PATH
    )


  
    # Convert To RGB
    

    image_rgb = image.convert(
        "RGB"
    )


    original_mode = image.mode


   
    # Resize Image
   

    resized_image = image_rgb.resize(

        MODEL_INPUT_SIZE,

        Image.Resampling.LANCZOS

    )


   
    # Convert To Array
   
    image_array = np.array(

        resized_image

    )


    
    # Normalize
   
    normalized_image = (

        image_array.astype(
            "float32"
        )
        /
        255.0

    )


   
    # Add Batch Dimension
   

    input_tensor = np.expand_dims(

        normalized_image,

        axis=0

    )


   
    # Tensor Validation
   

    expected_shape = (

        1,
        224,
        224,
        3

    )


    tensor_shape_valid = (

        input_tensor.shape ==
        expected_shape

    )


    value_range_valid = (

        input_tensor.min() >= 0.0

        and

        input_tensor.max() <= 1.0

    )


    preprocessing_status = (

        tensor_shape_valid

        and

        value_range_valid

    )


   
    # Store Preprocessing Metadata
   

    preprocessing_metadata = {


        "original_mode":

            original_mode,


        "converted_mode":

            "RGB",


        "target_size":

            list(MODEL_INPUT_SIZE),


        "tensor_shape":

            list(input_tensor.shape),


        "dtype":

            str(input_tensor.dtype),


        "minimum_value":

            float(input_tensor.min()),


        "maximum_value":

            float(input_tensor.max()),


        "normalization_range":

            list(NORMALIZATION_RANGE),


        "preprocessing_status":

            bool(preprocessing_status)

    }


 
    # Logging
   

    logger.info("-" * 42)


    logger.info(
        f"Original Image Mode       : {original_mode}"
    )


    logger.info(
        f"Converted Mode            : RGB"
    )


    logger.info(
        f"Resize Target             : {MODEL_INPUT_SIZE}"
    )


    logger.info(
        f"Tensor Shape              : {input_tensor.shape}"
    )


    logger.info(
        f"Tensor Data Type          : {input_tensor.dtype}"
    )


    logger.info(
        f"Minimum Tensor Value      : {input_tensor.min():.4f}"
    )


    logger.info(
        f"Maximum Tensor Value      : {input_tensor.max():.4f}"
    )


    logger.info(
        f"Shape Validation          : {tensor_shape_valid}"
    )


    logger.info(
        f"Value Range Validation    : {value_range_valid}"
    )


    logger.info("-" * 42)


    logger.info(
        f"Preprocessing Status      : {preprocessing_status}"
    )


    if not preprocessing_status:

        raise ValueError(
            "Image preprocessing validation failed"
        )


    logger.info(
        "Image Preprocessing Completed Successfully"
    )


    logger.info("=" * 70)



except Exception as e:


    logger.exception(
        f"Image preprocessing failed : {str(e)}"
    )

    raise
```

**9.Input Validation Test Cases**

```python


try:

    logger.info("=" * 70)
    logger.info("Running Input Validation Test Cases")
    logger.info("=" * 70)


    import os
    import json
    from datetime import datetime



    
    # Test Results Storage
   

    test_results = []



   
    # Helper Function
    
    def run_test_case(
        test_name,
        condition,
        expected_result
    ):

        result = {

            "test_case": test_name,

            "status": bool(condition),

            "expected": expected_result,

            "timestamp": str(datetime.now())

        }


        test_results.append(result)


        logger.info(
            f"{test_name:<45}: {condition}"
        )



   
    # Metadata Safe Extractor
    

    def check_metadata_status(
        metadata,
        keys
    ):


        if metadata is None:

            return False


        for key in keys:


            if key in metadata:

                value = metadata[key]


                if isinstance(value, bool):

                    return value



        return False



    
    # TC-01 Valid Input File
    

    valid_file_test = (

        "INPUT_IMAGE_PATH" in globals()

        and

        os.path.exists(INPUT_IMAGE_PATH)

    )


    run_test_case(

        "TC-01 Valid Input Image",

        valid_file_test,

        "PASS"

    )



   
    # TC-02 Image Format Validation
    
    supported_formats = [

        ".jpg",
        ".jpeg",
        ".png",
        ".bmp",
        ".tif",
        ".tiff"

    ]


    format_test = (

        os.path.splitext(

            INPUT_IMAGE_PATH

        )[1].lower()

        in

        supported_formats

    )



    run_test_case(

        "TC-02 Supported Image Format",

        format_test,

        "PASS"

    )



    
    # TC-03 Image Integrity
    

    integrity_test = check_metadata_status(

        globals().get(
            "image_integrity_metadata"
        ),

        [

            "integrity_status",

            "image_integrity_status"

        ]

    )



    run_test_case(

        "TC-03 Image Integrity",

        integrity_test,

        "PASS"

    )



   
    # TC-04 Channel Validation
    

    channel_test = check_metadata_status(

        globals().get(
            "image_channel_metadata"
        ),

        [

            "channel_validation",

            "channel_status",

            "valid_channels"

        ]

    )



    run_test_case(

        "TC-04 Channel Validation",

        channel_test,

        "PASS"

    )



    

    dimension_metadata = globals().get(

        "image_dimension_metadata"

    )


    dimension_test = False



    if dimension_metadata is not None:


        dimension_test = (

            dimension_metadata.get(

                "dimension_validation",

                False

            )

            or


            dimension_metadata.get(

                "resize_required",

                False

            )

            or


            (

                dimension_metadata.get(

                    "minimum_dimension_check",

                    False

                )

                and

                dimension_metadata.get(

                    "maximum_dimension_check",

                    False

                )

            )

        )



    run_test_case(

        "TC-05 Dimension Validation",

        dimension_test,

        "PASS"

    )



    
    # TC-06 Preprocessing Validation
   

    preprocessing_test = check_metadata_status(

        globals().get(

            "preprocessing_metadata"

        ),

        [

            "preprocessing_status"

        ]

    )



    run_test_case(

        "TC-06 MobileNetV2 Preprocessing",

        preprocessing_test,

        "PASS"

    )



    
    # Final Summary
   
    total_tests = len(test_results)


    passed_tests = sum(

        1

        for test in test_results

        if test["status"]

    )


    failed_tests = (

        total_tests

        -

        passed_tests

    )



    overall_status = (

        failed_tests == 0

    )



    test_summary = {


        "module":

            "Input Validation",


        "model":

            "RiceLeaf MobileNetV2",


        "execution_time":

            str(datetime.now()),


        "total_tests":

            total_tests,


        "passed_tests":

            passed_tests,


        "failed_tests":

            failed_tests,


        "overall_status":

            overall_status,


        "deployment_ready":

            overall_status,


        "test_results":

            test_results

    }



   
    # Save Test Report
   

    os.makedirs(

        INPUT_VALIDATION_DIR,

        exist_ok=True

    )


    test_report_path = os.path.join(

        INPUT_VALIDATION_DIR,

        "input_validation_test_report.json"

    )



    with open(

        test_report_path,

        "w"

    ) as file:


        json.dump(

            test_summary,

            file,

            indent=4

        )



   
    #  Logging

    

    logger.info("-" * 45)


    logger.info(

        f"Total Test Cases       : {total_tests}"

    )


    logger.info(

        f"Passed Tests           : {passed_tests}"

    )


    logger.info(

        f"Failed Tests           : {failed_tests}"

    )


    logger.info(

        f"Overall Status         : {overall_status}"

    )


    logger.info(

        f"Deployment Ready       : {overall_status}"

    )


    logger.info(

        f"Report Saved           : {os.path.exists(test_report_path)}"

    )


    logger.info(

        f"Report Path            : {test_report_path}"

    )


    logger.info("-" * 45)



    if not overall_status:


        raise ValueError(

            "Input Validation Test Cases Failed"

        )



    logger.info(

        "Input Validation Test Cases Completed Successfully"

    )


    logger.info("=" * 70)



except Exception as e:


    logger.exception(

        f"Input validation testing failed : {str(e)}"

    )

    raise
```

```python


try:

    logger.info("=" * 70)
    logger.info("Creating Safe Prediction Configuration")
    logger.info("=" * 70)


    import os
    import json



    
    # Verify Required Libraries
    
    logger.info(
        "Required libraries imported successfully"
    )



    
    # Base Directories
  

    if "ARTIFACTS_DIR" not in globals():

        ARTIFACTS_DIR = "artifacts"



    if "MODEL_DIR" not in globals():

        MODEL_DIR = os.path.join(

            ARTIFACTS_DIR,

            "model_registry"

        )



    # Prediction Directory

    PREDICTION_DIR = os.path.join(

        ARTIFACTS_DIR,

        "predictions"

    )


    os.makedirs(

        PREDICTION_DIR,

        exist_ok=True

    )



    
    # Model Configuration
  

    MODEL_NAME = (

        "RiceLeaf_MobileNetV2_TransferLearning"

    )


    MODEL_VERSION = "v1.0"



    MODEL_PATH = os.path.join(

        MODEL_DIR,

        "RiceLeaf_MobileNetV2_v1.0.keras"

    )



    CLASS_LABEL_PATH = os.path.join(

        MODEL_DIR,

        "RiceLeaf_MobileNetV2_v1.0_class_labels.json"

    )



    
    # Input Configuration
   

    IMAGE_SIZE = (

        224,

        224

    )


    INPUT_CHANNELS = 3


    INPUT_SHAPE = (

        1,

        224,

        224,

        3

    )



    NORMALIZATION_RANGE = [

        0.0,

        1.0

    ]



    
    # Prediction Configuration
   
    CONFIDENCE_THRESHOLD = 0.70


    TOP_K_PREDICTIONS = 3



    SUPPORTED_FORMATS = [

        ".jpg",

        ".jpeg",

        ".png",

        ".bmp",

        ".tif",

        ".tiff"

    ]



   
    # Prediction Output Configuration
    

    OUTPUT_FORMAT = {


        "class_name":

            True,


        "confidence":

            True,


        "probabilities":

            True,


        "timestamp":

            True

    }



    
    # Create Configuration Dictionary
  
    prediction_config = {


        "model_name":

            MODEL_NAME,


        "model_version":

            MODEL_VERSION,


        "model_path":

            MODEL_PATH,


        "class_label_path":

            CLASS_LABEL_PATH,


        "image_size":

            list(IMAGE_SIZE),


        "input_shape":

            list(INPUT_SHAPE),


        "input_channels":

            INPUT_CHANNELS,


        "normalization_range":

            NORMALIZATION_RANGE,


        "confidence_threshold":

            CONFIDENCE_THRESHOLD,


        "top_k_predictions":

            TOP_K_PREDICTIONS,


        "supported_formats":

            SUPPORTED_FORMATS,


        "output_format":

            OUTPUT_FORMAT

    }



    
    # Save Configuration
    

    config_path = os.path.join(

        PREDICTION_DIR,

        "prediction_config.json"

    )


    with open(

        config_path,

        "w"

    ) as file:


        json.dump(

            prediction_config,

            file,

            indent=4

        )



   
    # Validation Checks
   

    model_path_valid = os.path.exists(

        MODEL_PATH

    )


    label_path_valid = os.path.exists(

        CLASS_LABEL_PATH

    )


    directory_valid = os.path.exists(

        PREDICTION_DIR

    )


    image_size_valid = (

        IMAGE_SIZE == (224,224)

    )


    channels_valid = (

        INPUT_CHANNELS == 3

    )



    logger.info("-" * 45)


    logger.info(

        f"Prediction Directory : {PREDICTION_DIR}"

    )


    logger.info(

        f"Model Path           : {MODEL_PATH}"

    )


    logger.info(

        f"Class Label Path     : {CLASS_LABEL_PATH}"

    )


    logger.info(

        f"Image Size           : {IMAGE_SIZE}"

    )


    logger.info(

        f"Input Shape          : {INPUT_SHAPE}"

    )


    logger.info(

        f"Confidence Threshold : {CONFIDENCE_THRESHOLD}"

    )


    logger.info("-" * 45)


    logger.info(

        f"Model Available      : {model_path_valid}"

    )


    logger.info(

        f"Labels Available     : {label_path_valid}"

    )


    logger.info(

        f"Directory Created    : {directory_valid}"

    )


    logger.info(

        f"Image Size Valid     : {image_size_valid}"

    )


    logger.info(

        f"Channels Valid       : {channels_valid}"

    )


    logger.info("-" * 45)



    if not all([

        model_path_valid,

        label_path_valid,

        directory_valid,

        image_size_valid,

        channels_valid

    ]):


        raise ValueError(

            "Prediction Configuration Validation Failed"

        )



    logger.info(

        "Prediction Configuration Status : SUCCESS"

    )


    logger.info(

        f"Configuration Saved : {config_path}"

    )


    logger.info("=" * 70)



except Exception as e:


    logger.exception(

        f"Prediction configuration failed : {str(e)}"

    )

    raise
```

**2.Load Production Model**

```python


try:

    logger.info("=" * 70)
    logger.info("Loading Production MobileNetV2 Model")
    logger.info("=" * 70)



    import os
    import tensorflow as tf



    
    # Verify Configuration
   

    if "MODEL_PATH" not in globals():

        raise ValueError(
            "MODEL_PATH configuration not found"
        )



    logger.info(
        "Prediction configuration verified"
    )



  
    # Verify Model File
    
    model_file_exists = os.path.exists(

        MODEL_PATH

    )


    if not model_file_exists:


        raise FileNotFoundError(

            f"Model file not found: {MODEL_PATH}"

        )



    logger.info(

        "Production model file verified"

    )



  
   

    production_model = tf.keras.models.load_model(

        MODEL_PATH

    )



    logger.info(

        "Production model loaded successfully"

    )



    
    # Extract Model Details
    

    model_input_shape = (

        production_model.input_shape

    )


    model_output_shape = (

        production_model.output_shape

    )


    total_parameters = (

        production_model.count_params()

    )




    # Validation
   
    input_shape_valid = (

        model_input_shape == (None,224,224,3)

    )


    output_shape_valid = (

        model_output_shape[-1] == 3

    )



    parameters_valid = (

        total_parameters > 0

    )



   
    # Store Metadata
    

    production_model_metadata = {


        "model_path":

            MODEL_PATH,


        "input_shape":

            list(model_input_shape),


        "output_shape":

            list(model_output_shape),


        "total_parameters":

            total_parameters,


        "input_shape_valid":

            input_shape_valid,


        "output_shape_valid":

            output_shape_valid,


        "parameters_valid":

            parameters_valid

    }



  
    # Logging
    
    logger.info("-" * 45)


    logger.info(

        f"Model Input Shape       : {model_input_shape}"

    )


    logger.info(

        f"Model Output Shape      : {model_output_shape}"

    )


    logger.info(

        f"Total Parameters        : {total_parameters:,}"

    )


    logger.info("-" * 45)


    logger.info(

        f"Input Shape Valid       : {input_shape_valid}"

    )


    logger.info(

        f"Output Shape Valid      : {output_shape_valid}"

    )


    logger.info(

        f"Parameters Valid        : {parameters_valid}"

    )


    logger.info("-" * 45)



    if not all([

        input_shape_valid,

        output_shape_valid,

        parameters_valid

    ]):


        raise ValueError(

            "Production Model Validation Failed"

        )



    logger.info(

        "Production Model Status : SUCCESS"

    )


    logger.info(

        "Model Ready For Inference"

    )


    logger.info("=" * 70)



except Exception as e:


    logger.exception(

        f"Production model loading failed : {str(e)}"

    )

    raise
```

**3.Load Class Labels**

```python


try:

    logger.info("=" * 70)
    logger.info("Loading Production Class Labels")
    logger.info("=" * 70)



    import os
    import json



   
    # Verify Label Configuration
   

    if "CLASS_LABEL_PATH" not in globals():

        raise ValueError(

            "CLASS_LABEL_PATH not found"

        )



    logger.info(

        "Class label configuration verified"

    )



   
    # Validate Label File
  
    label_file_exists = os.path.exists(

        CLASS_LABEL_PATH

    )


    if not label_file_exists:


        raise FileNotFoundError(

            f"Class label file not found: {CLASS_LABEL_PATH}"

        )



    logger.info(

        "Class label file verified"

    )



  
    # Load JSON
    
    with open(

        CLASS_LABEL_PATH,

        "r"

    ) as file:


        class_labels_data = json.load(

            file

        )



    logger.info(

        "Class labels loaded successfully"

    )



    
    # Handle Label Format
    

    if isinstance(

        class_labels_data,

        dict

    ):


        class_indices = class_labels_data



    elif isinstance(

        class_labels_data,

        list

    ):


        class_indices = {


            str(index):

            label


            for index, label

            in enumerate(class_labels_data)

        }


    else:


        raise ValueError(

            "Unsupported class label format"

        )



   
    # Create Reverse Mapping
   
    index_to_class = {


        int(key):

        value


        for key, value

        in class_indices.items()

    }



   
    # Validation
 
    total_classes = len(

        index_to_class

    )


    class_count_valid = (

        total_classes == 3

    )


    index_validation = all(

        isinstance(

            key,

            int

        )

        for key in index_to_class.keys()

    )



    label_validation = all(

        isinstance(

            value,

            str

        )

        for value in index_to_class.values()

    )



   
    # Store Metadata
    
    class_label_metadata = {


        "label_path":

            CLASS_LABEL_PATH,


        "total_classes":

            total_classes,


        "index_to_class":

            index_to_class,


        "class_count_valid":

            class_count_valid,


        "index_validation":

            index_validation,


        "label_validation":

            label_validation

    }



   
    # Logging
   
    logger.info("-" * 45)


    logger.info(

        f"Class Mapping : {index_to_class}"

    )


    logger.info(

        f"Total Classes : {total_classes}"

    )


    logger.info("-" * 45)


    logger.info(

        f"Class Count Valid : {class_count_valid}"

    )


    logger.info(

        f"Index Validation  : {index_validation}"

    )


    logger.info(

        f"Label Validation  : {label_validation}"

    )


    logger.info("-" * 45)



    if not all([

        class_count_valid,

        index_validation,

        label_validation

    ]):


        raise ValueError(

            "Class Label Validation Failed"

        )



    logger.info(

        "Class Labels Status : SUCCESS"

    )


    logger.info(

        "Prediction Labels Ready"

    )


    logger.info("=" * 70)



except Exception as e:


    logger.exception(

        f"Class label loading failed : {str(e)}"

    )

    raise
```

**4.Validate Prediction Input**

```python


try:

    logger.info("=" * 70)
    logger.info("Validating Prediction Input Before Inference")
    logger.info("=" * 70)



    import os
    import numpy as np



   
    # Verify Required Objects
    

    required_objects = [

        "INPUT_IMAGE_PATH",

        "production_model",

        "IMAGE_SIZE",

        "INPUT_CHANNELS"

    ]



    missing_objects = [

        obj

        for obj in required_objects

        if obj not in globals()

    ]



    if missing_objects:


        raise ValueError(

            f"Missing required objects: {missing_objects}"

        )



    logger.info(

        "Required prediction objects verified"

    )



   
    # File Validation
    

    file_exists = os.path.exists(

        INPUT_IMAGE_PATH

    )


    file_not_empty = (

        os.path.getsize(

            INPUT_IMAGE_PATH

        ) > 0

    )



    
    # Format Validation
 

    extension = os.path.splitext(

        INPUT_IMAGE_PATH

    )[1].lower()



    format_valid = (

        extension in SUPPORTED_FORMATS

    )



   
    # Image Loading Validation
   

    image_load_valid = False

    image_shape = None



    try:


        from PIL import Image


        image = Image.open(

            INPUT_IMAGE_PATH

        )


        image.verify()


        image_load_valid = True



    except Exception:


        image_load_valid = False



    
    # RGB Validation
    

    rgb_valid = False



    if image_load_valid:


        image = Image.open(

            INPUT_IMAGE_PATH

        ).convert(

            "RGB"

        )


        image_array = np.array(

            image

        )


        image_shape = image_array.shape



        rgb_valid = (

            len(image_shape) == 3

            and

            image_shape[-1] == 3

        )



   
    # Model Compatibility
    
    model_input_valid = (

        IMAGE_SIZE == (224,224)

        and

        INPUT_CHANNELS == 3

    )



   
    
    validation_status = {


        "file_exists":

            file_exists,


        "file_not_empty":

            file_not_empty,


        "format_valid":

            format_valid,


        "image_readable":

            image_load_valid,


        "rgb_valid":

            rgb_valid,


        "model_input_valid":

            model_input_valid

    }



    prediction_input_valid = all(

        validation_status.values()

    )



   
    # Store Metadata
  
    prediction_input_metadata = {


        "input_path":

            INPUT_IMAGE_PATH,


        "image_shape":

            list(image_shape)

            if image_shape

            else None,


        "validation_status":

            validation_status,


        "prediction_input_valid":

            prediction_input_valid

    }



    
    # Logging
    

    logger.info("-" * 45)


    logger.info(

        f"Input Image       : {INPUT_IMAGE_PATH}"

    )


    logger.info(

        f"Image Shape       : {image_shape}"

    )


    logger.info("-" * 45)



    for key, value in validation_status.items():


        logger.info(

            f"{key:<25}: {value}"

        )



    logger.info("-" * 45)


    logger.info(

        f"Prediction Input Valid : {prediction_input_valid}"

    )



    if not prediction_input_valid:


        raise ValueError(

            "Prediction Input Validation Failed"

        )



    logger.info(

        "Prediction Input Validation : SUCCESS"

    )


    logger.info(

        "Image Ready For Preprocessing"

    )


    logger.info("=" * 70)



except Exception as e:


    logger.exception(

        f"Prediction input validation failed : {str(e)}"

    )

    raise
```

**5.Generate Prediction**

```python


try:

    logger.info("=" * 70)
    logger.info("Generating MobileNetV2 Prediction")
    logger.info("=" * 70)



    import numpy as np
    from PIL import Image
    import tensorflow as tf



    
    # Verify Required Objects
  

    required_objects = [

        "production_model",

        "index_to_class",

        "INPUT_IMAGE_PATH",

        "IMAGE_SIZE"

    ]


    missing_objects = [

        obj

        for obj in required_objects

        if obj not in globals()

    ]



    if missing_objects:


        raise ValueError(

            f"Missing required objects: {missing_objects}"

        )



    logger.info(

        "Prediction objects verified"

    )



   
    # Load Image
    

    image = Image.open(

        INPUT_IMAGE_PATH

    ).convert(

        "RGB"

    )



    logger.info(

        "Image loaded successfully"

    )



    
    # Preprocessing
   

    image_resized = image.resize(

        IMAGE_SIZE

    )



    image_array = np.array(

        image_resized

    )



    image_array = image_array.astype(

        "float32"

    )



    image_array = image_array / 255.0



    input_tensor = np.expand_dims(

        image_array,

        axis=0

    )



    logger.info(

        f"Input Tensor Shape : {input_tensor.shape}"

    )



    
    # Model Prediction
   
    predictions = production_model.predict(

        input_tensor,

        verbose=0

    )



    probabilities = predictions[0]



    predicted_index = int(

        np.argmax(

            probabilities

        )

    )



    confidence = float(

        np.max(

            probabilities

        )

    )



    predicted_class = index_to_class[

        predicted_index

    ]



    
    # Top-K Predictions
  
    top_k_indices = np.argsort(

        probabilities

    )[::-1][:TOP_K_PREDICTIONS]



    top_k_predictions = []



    for index in top_k_indices:


        top_k_predictions.append(

            {


                "class_index":

                    int(index),


                "class_name":

                    index_to_class[int(index)],


                "confidence":

                    float(probabilities[index])

            }

        )



   
    # Store Prediction Metadata
    

    prediction_result = {


        "input_image":

            INPUT_IMAGE_PATH,


        "predicted_index":

            predicted_index,


        "predicted_class":

            predicted_class,


        "confidence":

            confidence,


        "probabilities":

            probabilities.tolist(),


        "top_k_predictions":

            top_k_predictions

    }



    
    # Logging
   

    logger.info("-" * 45)


    logger.info(

        f"Predicted Index : {predicted_index}"

    )


    logger.info(

        f"Predicted Class : {predicted_class}"

    )


    logger.info(

        f"Confidence      : {confidence:.4f}"

    )


    logger.info("-" * 45)



    logger.info(

        "Top Predictions"

    )



    for item in top_k_predictions:


        logger.info(

            f"{item['class_name']} : {item['confidence']:.4f}"

        )



    logger.info("-" * 45)



    logger.info(

        "Prediction Generation Status : SUCCESS"

    )


    logger.info("=" * 70)



except Exception as e:


    logger.exception(

        f"Prediction generation failed : {str(e)}"

    )

    raise
```

**6.Confidence Analysis**

```python


try:

    logger.info("=" * 70)
    logger.info("Running Prediction Confidence Analysis")
    logger.info("=" * 70)



   
    # Verify Required Objects
    

    required_objects = [

        "prediction_result",

        "CONFIDENCE_THRESHOLD"

    ]


    missing_objects = [

        obj

        for obj in required_objects

        if obj not in globals()

    ]



    if missing_objects:


        raise ValueError(

            f"Missing required objects: {missing_objects}"

        )



    logger.info(

        "Confidence analysis objects verified"

    )



   
    # Extract Confidence
   

    confidence_score = (

        prediction_result["confidence"]

    )



    predicted_class = (

        prediction_result["predicted_class"]

    )



    
    # Confidence Classification
    

    if confidence_score >= 0.85:


        confidence_level = "HIGH"



    elif confidence_score >= 0.70:


        confidence_level = "MEDIUM"



    else:


        confidence_level = "LOW"



    
    # Reliability Decision
    

    prediction_reliable = (

        confidence_score >= CONFIDENCE_THRESHOLD

    )



    if prediction_reliable:


        reliability_status = "RELIABLE"


    else:


        reliability_status = "UNCERTAIN"



  
  
    if confidence_level == "HIGH":


        recommendation = (

            "Prediction can be accepted"

        )



    elif confidence_level == "MEDIUM":


        recommendation = (

            "Prediction acceptable with monitoring"

        )



    else:


        recommendation = (

            "Manual verification recommended"

        )



    
    # Store Confidence Metadata
   
    confidence_analysis = {


        "predicted_class":

            predicted_class,


        "confidence_score":

            float(confidence_score),


        "confidence_percentage":

            round(

                confidence_score * 100,

                2

            ),


        "confidence_threshold":

            CONFIDENCE_THRESHOLD,


        "confidence_level":

            confidence_level,


        "prediction_reliable":

            prediction_reliable,


        "reliability_status":

            reliability_status,


        "recommendation":

            recommendation

    }



    
    # Logging
   

    logger.info("-" * 45)


    logger.info(

        f"Predicted Class       : {predicted_class}"

    )


    logger.info(

        f"Confidence Score      : {confidence_score:.4f}"

    )


    logger.info(

        f"Confidence Percentage : {confidence_score*100:.2f}%"

    )


    logger.info(

        f"Threshold             : {CONFIDENCE_THRESHOLD}"

    )


    logger.info("-" * 45)


    logger.info(

        f"Confidence Level      : {confidence_level}"

    )


    logger.info(

        f"Reliability Status    : {reliability_status}"

    )


    logger.info(

        f"Recommendation        : {recommendation}"

    )


    logger.info("-" * 45)



    logger.info(

        "Confidence Analysis Status : SUCCESS"

    )


    logger.info("=" * 70)



except Exception as e:


    logger.exception(

        f"Confidence analysis failed : {str(e)}"

    )

    raise
```

**7.Prediction Response Format**

```python
try:

    logger.info("=" * 70)
    logger.info("Creating Production Prediction Response Format")
    logger.info("=" * 70)


    # Imports
    from datetime import datetime
    import json
    import os



    # Verify Required Objects

    required_objects = [

        "prediction_result",
        "confidence_analysis",
        "MODEL_NAME",
        "MODEL_VERSION"

    ]


    missing_objects = [

        obj
        for obj in required_objects
        if obj not in globals()

    ]


    if missing_objects:

        raise ValueError(
            f"Missing required objects: {missing_objects}"
        )


    logger.info(
        "Prediction response objects verified"
    )



    # Create Production Response Object

    prediction_response = {


        "model": {


            "name":

                MODEL_NAME,


            "version":

                MODEL_VERSION

        },



        "prediction": {


            "class_index":

                prediction_result["predicted_index"],



            "class_name":

                prediction_result["predicted_class"],



            "confidence":

                round(

                    prediction_result["confidence"],

                    4

                ),



            "confidence_percentage":

                round(

                    prediction_result["confidence"] * 100,

                    2

                )

        },



        "confidence_analysis": {


            "level":

                confidence_analysis["confidence_level"],



            "status":

                confidence_analysis["reliability_status"],



            "threshold":

                confidence_analysis["confidence_threshold"],



            "recommendation":

                confidence_analysis["recommendation"]

        },



        "top_predictions":

            prediction_result["top_k_predictions"],




        "input": {


            "image":

                prediction_result["input_image"]

        },



        "timestamp":

            datetime.now().isoformat()

    }




    # Save Response JSON


    response_path = os.path.join(

        PREDICTION_DIR,

        "prediction_response.json"

    )



    with open(

        response_path,

        "w"

    ) as file:


        json.dump(

            prediction_response,

            file,

            indent=4

        )




    # Response Validation


    required_fields = [

        "model",

        "prediction",

        "confidence_analysis",

        "top_predictions",

        "timestamp"

    ]



    response_valid = all(

        field in prediction_response

        for field in required_fields

    )



    file_saved = os.path.exists(

        response_path

    )





    # Logging


    logger.info("-" * 45)


    logger.info(

        f"Predicted Class : {prediction_response['prediction']['class_name']}"

    )


    logger.info(

        f"Confidence      : {prediction_response['prediction']['confidence_percentage']}%"

    )


    logger.info(

        f"Status          : {prediction_response['confidence_analysis']['status']}"

    )


    logger.info("-" * 45)



    logger.info(

        f"Response Valid  : {response_valid}"

    )


    logger.info(

        f"JSON Saved      : {file_saved}"

    )


    logger.info(

        f"Response Path   : {response_path}"

    )


    logger.info("-" * 45)





    if not all([

        response_valid,

        file_saved

    ]):


        raise ValueError(

            "Prediction Response Validation Failed"

        )




    logger.info(

        "Prediction Response Format : SUCCESS"

    )


    logger.info(

        "API Ready Response Generated"

    )


    logger.info("=" * 70)



except Exception as e:


    logger.exception(

        f"Prediction response creation failed : {str(e)}"

    )

    raise
```

```python
try:

    logger.info("=" * 70)
    logger.info("Saving Prediction History Log")
    logger.info("=" * 70)



    # Imports

    import os
    import json
    from datetime import datetime



    # Verify Required Objects


    required_objects = [

        "prediction_response",
        "PREDICTION_DIR"

    ]



    missing_objects = [

        obj

        for obj in required_objects

        if obj not in globals()

    ]



    if missing_objects:


        raise ValueError(

            f"Missing required objects: {missing_objects}"

        )



    logger.info(

        "Prediction logging objects verified"

    )




    # Prediction Log Path


    prediction_log_path = os.path.join(

        PREDICTION_DIR,

        "prediction_history.json"

    )





    # Load Existing History


    if os.path.exists(

        prediction_log_path

    ):


        with open(

            prediction_log_path,

            "r"

        ) as file:


            prediction_history = json.load(

                file

            )


    else:


        prediction_history = []





    # Create Log Entry


    log_entry = {


        "prediction_id":

            len(prediction_history) + 1,



        "timestamp":

            datetime.now().isoformat(),



        "model_name":

            prediction_response["model"]["name"],



        "model_version":

            prediction_response["model"]["version"],



        "input_image":

            prediction_response["input"]["image"],



        "predicted_class":

            prediction_response["prediction"]["class_name"],



        "confidence":

            prediction_response["prediction"]["confidence"],



        "confidence_percentage":

            prediction_response["prediction"]["confidence_percentage"],



        "status":

            prediction_response["confidence_analysis"]["status"]

    }




    # Append History


    prediction_history.append(

        log_entry

    )





    # Save Prediction History


    with open(

        prediction_log_path,

        "w"

    ) as file:


        json.dump(

            prediction_history,

            file,

            indent=4

        )





    # Validation


    log_saved = os.path.exists(

        prediction_log_path

    )



    log_count = len(

        prediction_history

    )



    required_fields = [

        "prediction_id",

        "timestamp",

        "predicted_class",

        "confidence",

        "status"

    ]



    log_structure_valid = all(

        field in log_entry

        for field in required_fields

    )





    # Logging


    logger.info("-" * 45)


    logger.info(

        f"Prediction ID       : {log_entry['prediction_id']}"

    )


    logger.info(

        f"Predicted Class     : {log_entry['predicted_class']}"

    )


    logger.info(

        f"Confidence          : {log_entry['confidence_percentage']}%"

    )


    logger.info(

        f"Status              : {log_entry['status']}"

    )


    logger.info("-" * 45)



    logger.info(

        f"Log Saved           : {log_saved}"

    )


    logger.info(

        f"Total Predictions   : {log_count}"

    )


    logger.info(

        f"Log Path            : {prediction_log_path}"

    )


    logger.info("-" * 45)





    if not all([

        log_saved,

        log_structure_valid

    ]):


        raise ValueError(

            "Prediction Logging Validation Failed"

        )





    logger.info(

        "Prediction Logging Status : SUCCESS"

    )


    logger.info(

        "Prediction History Updated"

    )


    logger.info("=" * 70)




except Exception as e:


    logger.exception(

        f"Prediction logging failed : {str(e)}"

    )

    raise
```

**5.Create Image Prediction Endpoint**

```python


try:

    logger.info("=" * 70)
    logger.info("Creating FastAPI Image Prediction Endpoint")
    logger.info("=" * 70)

    import io
    import numpy as np

    from datetime import datetime

    from fastapi import UploadFile, File, HTTPException
    from PIL import Image


   
    # Verify Required Objects
   

    required_objects = [

        "app",
        "production_model",
        "class_labels"

    ]

    missing_objects = [

        obj

        for obj in required_objects

        if obj not in globals()

    ]

    if missing_objects:

        raise ValueError(

            f"Missing required objects: {missing_objects}"

        )

    logger.info(
        "Prediction endpoint objects verified"
    )


    
    # Prediction Endpoint
  

    @app.post("/predict")

    async def predict_image(

        file: UploadFile = File(...)

    ):

        try:

            
            # Validate File
            

            filename = file.filename.lower()

            if not filename.endswith(

                tuple(SUPPORTED_API_FORMATS)

            ):

                raise HTTPException(

                    status_code=400,

                    detail="Unsupported image format"

                )


            
            # Read Image
            

            image_bytes = await file.read()

            image = Image.open(

                io.BytesIO(image_bytes)

            )

            image = image.convert("RGB")


            
            # Preprocess Image
           

            image = image.resize(

                IMAGE_SIZE

            )

            image_array = np.array(

                image,

                dtype=np.float32

            )

            image_array = image_array / 255.0

            image_array = np.expand_dims(

                image_array,

                axis=0

            )


           
            # Prediction
            

            predictions = production_model.predict(

                image_array,

                verbose=0

            )

            predicted_index = int(

                np.argmax(

                    predictions[0]

                )

            )

            confidence = float(

                np.max(

                    predictions[0]

                )

            )

            predicted_class = class_labels[

                str(predicted_index)

            ]


           
            # Top Predictions
           

            top_predictions = {}

            for index, probability in enumerate(predictions[0]):

                top_predictions[

                    class_labels[str(index)]

                ] = round(

                    float(probability),

                    4

                )


           
            # Response
           
            response = {

                "status": "success",

                "prediction": {

                    "class_index": predicted_index,

                    "class_name": predicted_class,

                    "confidence": round(

                        confidence,

                        4

                    ),

                    "confidence_percentage": round(

                        confidence * 100,

                        2

                    )

                },

                "top_predictions": top_predictions,

                "model": {

                    "name": "RiceLeaf_MobileNetV2_TransferLearning",

                    "version": "v1.0"

                },

                "timestamp": datetime.now().isoformat()

            }

            return response


        
        # Preserve HTTP Errors
        

        except HTTPException:

            raise


        
        # Unexpected Errors
        

        except Exception as error:

            logger.exception(

                f"Prediction Endpoint Error : {str(error)}"

            )

            raise HTTPException(

                status_code=500,

                detail=str(error)

            )


    
    # Validation
  

    endpoint_created = any(

        route.path == "/predict"

        for route in app.routes

    )

    logger.info("-" * 45)

    logger.info(

        f"Endpoint Created : {endpoint_created}"

    )

    logger.info(

        "Endpoint Path    : /predict"

    )

    logger.info(

        "HTTP Method      : POST"

    )

    logger.info("-" * 45)

    if not endpoint_created:

        raise ValueError(

            "Prediction endpoint creation failed"

        )

    logger.info(

        "Image Prediction Endpoint : SUCCESS"

    )

    logger.info(

        "API Ready For Image Inference"

    )

    logger.info("=" * 70)

except Exception as e:

    logger.exception(

        f"Prediction endpoint creation failed : {str(e)}"

    )

    raise
```

**6.API Input Validation**

```python


try:

    logger.info("=" * 70)
    logger.info("Adding Production API Input Validation")
    logger.info("=" * 70)



    import io
    import numpy as np
    from PIL import Image
    from fastapi import UploadFile, File, HTTPException



   
    # API Validation Configuration
   

    MAX_FILE_SIZE_MB = 10

    MAX_FILE_SIZE_BYTES = (

        MAX_FILE_SIZE_MB * 1024 * 1024

    )


    API_SUPPORTED_FORMATS = [

        ".jpg",

        ".jpeg",

        ".png",

        ".bmp",

        ".tif",

        ".tiff"

    ]



    logger.info(

        "Validation configuration loaded"

    )



  
    # Enhanced Prediction Endpoint
   

    @app.post("/predict")


    async def predict_image_validated(

        file: UploadFile = File(...)

    ):


        try:


           
            # 1. File Validation
         

            if file.filename is None:


                raise HTTPException(

                    status_code=400,

                    detail="Filename missing"

                )



            filename = file.filename.lower()



            extension = (

                "." + filename.split(".")[-1]

            )



            if extension not in API_SUPPORTED_FORMATS:


                raise HTTPException(

                    status_code=400,

                    detail="Unsupported image format"

                )



           
            # 2. Read Uploaded File
          

            image_bytes = await file.read()



            if len(image_bytes) == 0:


                raise HTTPException(

                    status_code=400,

                    detail="Empty image file"

                )



            # File size validation

            if len(image_bytes) > MAX_FILE_SIZE_BYTES:


                raise HTTPException(

                    status_code=400,

                    detail="Image file size exceeds limit"

                )



           
            # 3. Image Integrity Validation
           


            try:


                image = Image.open(

                    io.BytesIO(image_bytes)

                )


                image.verify()



            except Exception:


                raise HTTPException(

                    status_code=400,

                    detail="Corrupted image file"

                )



            # Reload image after verify

            image = Image.open(

                io.BytesIO(image_bytes)

            )



           
            # 4. Channel Validation
          


            original_mode = image.mode



            image = image.convert(

                "RGB"

            )



            channels = len(

                image.getbands()

            )



            if channels != 3:


                raise HTTPException(

                    status_code=400,

                    detail="Invalid image channels"

                )



           
            # 5. Preprocessing
          


            image = image.resize(

                IMAGE_SIZE

            )



            image_array = np.array(

                image

            )



            image_array = (

                image_array / 255.0

            )



            image_array = np.expand_dims(

                image_array,

                axis=0

            )



            
            # 6. Prediction
           

            prediction = production_model.predict(

                image_array,

                verbose=0

            )



            class_index = int(

                np.argmax(

                    prediction[0]

                )

            )



            confidence = float(

                np.max(

                    prediction[0]

                )

            )



            class_name = class_labels[

                str(class_index)

            ]



            
            # Response
           

            return {


                "status":

                    "success",


                "validation":

                {


                    "format_valid":

                        True,


                    "integrity_valid":

                        True,


                    "channels":

                        channels

                },


                "prediction":

                {


                    "class_index":

                        class_index,


                    "class_name":

                        class_name,


                    "confidence":

                        round(

                            confidence,

                            4

                        )

                },


                "model":

                {

                    "name":

                    "RiceLeaf_MobileNetV2_TransferLearning",


                    "version":

                    "v1.0"

                }

            }



        except HTTPException as error:


            raise error



        except Exception as error:


            raise HTTPException(

                status_code=500,

                detail=str(error)

            )



   
    # Validation
   


    endpoint_exists = any(

        route.path == "/predict"

        for route in app.routes

    )



    logger.info("-" * 45)


    logger.info(

        f"Prediction API Available : {endpoint_exists}"

    )


    logger.info(

        f"Max File Size           : {MAX_FILE_SIZE_MB} MB"

    )


    logger.info(

        f"Supported Formats       : {API_SUPPORTED_FORMATS}"

    )


    logger.info("-" * 45)



    logger.info(

        "API Input Validation Added Successfully"

    )


    logger.info("=" * 70)



except Exception as e:


    logger.exception(

        f"API input validation failed : {str(e)}"

    )

    raise
```

**7.API Response Schema & Documentation**

```python


try:

    logger.info("=" * 70)
    logger.info("Creating API Response Schema Documentation")
    logger.info("=" * 70)



    from pydantic import BaseModel, Field
    from typing import Dict, Any



   
    # Response Schema Definitions
  

    class PredictionResult(BaseModel):


        class_index: int = Field(

            description="Predicted disease class index"

        )


        class_name: str = Field(

            description="Predicted rice leaf disease name"

        )


        confidence: float = Field(

            description="Prediction confidence score"

        )



    class ValidationResult(BaseModel):


        format_valid: bool = Field(

            description="Image format validation status"

        )


        integrity_valid: bool = Field(

            description="Image integrity validation status"

        )


        channels: int = Field(

            description="Number of image channels"

        )



    class ModelInformation(BaseModel):


        name: str = Field(

            description="ML model name"

        )


        version: str = Field(

            description="Model version"

        )



    class PredictionResponse(BaseModel):


        status: str = Field(

            description="API execution status"

        )


        validation: ValidationResult


        prediction: PredictionResult


        model: ModelInformation



    logger.info(

        "Response schemas created successfully"

    )



   
    # FastAPI Metadata Update
   

    app.title = (

        "Rice Leaf Disease Classification API"

    )


    app.description = (

        "Production API for Rice Leaf Disease Detection "

        "using MobileNetV2 Transfer Learning"

    )


    app.version = "1.0.0"



    logger.info(

        "API metadata configured"

    )



    
    # Update Predict Endpoint Documentation
    


    for route in app.routes:


        if route.path == "/predict":


            route.response_model = PredictionResponse



    logger.info(

        "Prediction endpoint schema attached"

    )



   
    # Documentation Metadata
   

    api_metadata = {


        "API_Name":

            app.title,


        "Version":

            app.version,


        "Model":

            "RiceLeaf_MobileNetV2_TransferLearning",


        "Model_Version":

            "v1.0",


        "Endpoint":

            "/predict",


        "Method":

            "POST",


        "Documentation":

            "/docs"

    }



    import os
    import json



    FASTAPI_DIR = os.path.join(

        "artifacts",

        "fastapi"

    )


    os.makedirs(

        FASTAPI_DIR,

        exist_ok=True

    )



    metadata_path = os.path.join(

        FASTAPI_DIR,

        "api_response_schema_metadata.json"

    )



    with open(

        metadata_path,

        "w"

    ) as file:


        json.dump(

            api_metadata,

            file,

            indent=4

        )



    
    # Validation
    


    schema_created = True


    metadata_saved = os.path.exists(

        metadata_path

    )


    docs_available = (

        app.docs_url == "/docs"

    )



    logger.info("-" * 45)


    logger.info(

        f"Schema Created       : {schema_created}"

    )


    logger.info(

        f"Metadata Saved       : {metadata_saved}"

    )


    logger.info(

        f"Swagger Docs Enabled : {docs_available}"

    )


    logger.info(

        f"Documentation URL    : {app.docs_url}"

    )


    logger.info("-" * 45)



    if not all([

        schema_created,

        metadata_saved,

        docs_available

    ]):


        raise ValueError(

            "API documentation setup failed"

        )



    logger.info(

        "API Response Schema Documentation : SUCCESS"

    )


    logger.info(

        "Swagger Documentation Ready"

    )


    logger.info("=" * 70)



except Exception as e:


    logger.exception(

        f"API schema documentation failed : {str(e)}"

    )

    raise
```

**Production Prediction Endpoint**

```python

# 24.5 Production Prediction Endpoint (Fixed)


try:

    logger.info("=" * 70)
    logger.info("Creating Production Prediction Endpoint")
    logger.info("=" * 70)


    from fastapi import UploadFile, File, HTTPException
    from PIL import Image
    import numpy as np
    import io
    import tensorflow as tf



    @app.post(

        "/predict",

        tags=["Prediction"]

    )


    async def predict(

        file: UploadFile = File(...)

    ):


        try:


            
            # File Validation
           


            if not file.filename:


                raise HTTPException(

                    status_code=400,

                    detail="Filename missing"

                )



            allowed_formats = [

                ".jpg",

                ".jpeg",

                ".png",

                ".bmp"

            ]



            extension = (

                "." +

                file.filename.split(".")[-1].lower()

            )



            if extension not in allowed_formats:


                raise HTTPException(

                    status_code=400,

                    detail="Unsupported image format"

                )



           
            # Read Image
          

            image_bytes = await file.read()



            image = Image.open(

                io.BytesIO(image_bytes)

            )



            image = image.convert(

                "RGB"

            )



            original_size = image.size



            
            # Preprocessing
           


            image = image.resize(

                (224,224)

            )



            image_array = np.array(

                image

            )



            image_array = (

                image_array / 255.0

            )



            input_tensor = np.expand_dims(

                image_array,

                axis=0

            )



            
            # Prediction
            

            predictions = production_model.predict(

                input_tensor,

                verbose=0

            )



            predicted_index = int(

                np.argmax(

                    predictions[0]

                )

            )



            confidence = float(

                np.max(

                    predictions[0]

                )

            )



            predicted_class = class_labels[

                str(predicted_index)

            ]



            
            # Top Predictions
            


            top_predictions = {}



            for idx, score in enumerate(predictions[0]):


                top_predictions[

                    class_labels[str(idx)]

                ] = round(

                    float(score),

                    4

                )



            
            # Response
           


            return {


                "status":

                    "success",



                "validation":

                {


                    "format_valid":

                        True,


                    "integrity_valid":

                        True,


                    "channels":

                        3

                },



                "prediction":

                {


                    "class_index":

                        predicted_index,


                    "class_name":

                        predicted_class,


                    "confidence":

                        round(

                            confidence,

                            4

                        )

                },



                "top_predictions":

                    top_predictions,



                "image":

                {


                    "filename":

                        file.filename,


                    "original_size":

                        original_size

                },



                "model":

                {


                    "name":

                    "RiceLeaf_MobileNetV2_TransferLearning",


                    "version":

                    "v1.0"

                }


            }



        except Exception as e:


            logger.exception(

                f"Prediction failed : {str(e)}"

            )


            raise HTTPException(

                status_code=500,

                detail=str(e)

            )



    logger.info(

        "Prediction endpoint created successfully"

    )


    logger.info("=" * 70)



except Exception as e:


    logger.exception(

        f"Prediction endpoint creation failed : {str(e)}"

    )

    raise
```

```python
# =============================================================================
# 24.10 API Testing
# =============================================================================

try:

    logger.info("=" * 70)
    logger.info("Running FastAPI Endpoint Testing")
    logger.info("=" * 70)

    import os
    import json

    from fastapi.testclient import TestClient

    # -------------------------------------------------------------------------
    # Verify Required Objects
    # -------------------------------------------------------------------------

    required_objects = [

        "app",
        "INPUT_IMAGE_PATH"

    ]

    missing_objects = [

        obj

        for obj in required_objects

        if obj not in globals()

    ]

    if missing_objects:

        raise ValueError(

            f"Missing required objects : {missing_objects}"

        )

    logger.info(
        "Required objects verified"
    )

    # -------------------------------------------------------------------------
    # Create Test Client
    # -------------------------------------------------------------------------

    client = TestClient(app)

    logger.info(
        "FastAPI Test Client Created"
    )

    # -------------------------------------------------------------------------
    # Test Results
    # -------------------------------------------------------------------------

    test_results = {}

    # -------------------------------------------------------------------------
    # TC-01 Health Endpoint
    # -------------------------------------------------------------------------

    response = client.get("/health")

    tc1 = response.status_code == 200

    test_results["Health Endpoint"] = tc1

    logger.info(
        f"TC-01 Health Endpoint              : {tc1}"
    )

    # -------------------------------------------------------------------------
    # TC-02 Prediction Endpoint
    # -------------------------------------------------------------------------

    with open(INPUT_IMAGE_PATH, "rb") as image_file:

        response = client.post(

            "/predict",

            files={

                "file": (

                    os.path.basename(INPUT_IMAGE_PATH),

                    image_file,

                    "image/jpeg"

                )

            }

        )

    tc2 = response.status_code == 200

    test_results["Prediction Endpoint"] = tc2

    logger.info(
        f"TC-02 Prediction Endpoint          : {tc2}"
    )

    # -------------------------------------------------------------------------
    # TC-03 Response Schema
    # -------------------------------------------------------------------------

    tc3 = False

    if tc2:

        data = response.json()

        required_keys = [

            "status",
            "prediction",
            "model",
            "timestamp"

        ]

        tc3 = all(

            key in data

            for key in required_keys

        )

    test_results["Response Schema"] = tc3

    logger.info(
        f"TC-03 Response Schema              : {tc3}"
    )

    # -------------------------------------------------------------------------
    # TC-04 Invalid File
    # -------------------------------------------------------------------------

    response = client.post(

        "/predict",

        files={

            "file": (

                "dummy.txt",

                b"Hello World",

                "text/plain"

            )

        }

    )

    tc4 = response.status_code == 400

    test_results["Invalid File Handling"] = tc4

    logger.info(
        f"TC-04 Invalid File Handling        : {tc4}"
    )

    # -------------------------------------------------------------------------
    # Summary
    # -------------------------------------------------------------------------

    total_tests = len(test_results)

    passed_tests = sum(test_results.values())

    failed_tests = total_tests - passed_tests

    overall_status = passed_tests == total_tests

    logger.info("-" * 45)

    logger.info(
        f"Total Tests          : {total_tests}"
    )

    logger.info(
        f"Passed Tests         : {passed_tests}"
    )

    logger.info(
        f"Failed Tests         : {failed_tests}"
    )

    logger.info(
        f"Deployment Ready     : {overall_status}"
    )

    # -------------------------------------------------------------------------
    # Save Report
    # -------------------------------------------------------------------------

    report = {

        "total_tests": total_tests,
        "passed_tests": passed_tests,
        "failed_tests": failed_tests,
        "overall_status": overall_status,
        "results": test_results

    }

    report_path = os.path.join(

        FASTAPI_DIR,

        "fastapi_test_report.json"

    )

    with open(

        report_path,

        "w"

    ) as file:

        json.dump(

            report,

            file,

            indent=4

        )

    logger.info(
        f"Report Saved : {report_path}"
    )

    logger.info("=" * 70)

    if overall_status:

        logger.info(
            "FastAPI Endpoint Testing : SUCCESS"
        )

    else:

        logger.warning(
            "FastAPI Endpoint Testing : FAILED"
        )

except Exception as e:

    logger.exception(
        f"FastAPI testing failed : {str(e)}"
    )

    raise
```
